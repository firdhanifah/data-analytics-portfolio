## 0. Setup — Temukan Workspace Starter

Simpan notebook idealnya di `notebooks/sesi_01/`.

Jika auto-detect gagal, isi `PROJECT_ROOT_OVERRIDE`.

```python
print("hello world")
```

```python
from __future__ import annotations

from pathlib import Path
from datetime import datetime
import json
import platform
import sys

import pandas as pd

PROJECT_ROOT_OVERRIDE = None

REQUIRED_DIRS = [
    "data/raw",
    "data/processed",
    "notebooks",
    "sql",
    "outputs",
    "docs",
    "practice",
    "utils",
]

def resolve_project_root(override=None) -> Path:
    if override is not None:
        root = Path(override).expanduser().resolve()
        if not (root / "README.md").exists():
            raise FileNotFoundError(
                f"PROJECT_ROOT_OVERRIDE tidak valid: {root}\n"
                "README.md Workspace Starter tidak ditemukan."
            )
        return root

    start = Path.cwd().resolve()
    candidates = [start, *start.parents]

    for candidate in candidates:
        if (candidate / "README.md").exists() and (candidate / "data").exists():
            return candidate

    for candidate in candidates:
        nested = candidate / "Bootcamp_Data_Analyst"
        if (nested / "README.md").exists() and (nested / "data").exists():
            return nested.resolve()

    raise FileNotFoundError(
        "Workspace Starter tidak ditemukan. "
        "Buka folder Bootcamp_Data_Analyst sebagai project "
        "atau isi PROJECT_ROOT_OVERRIDE."
    )

ROOT = resolve_project_root(PROJECT_ROOT_OVERRIDE)
OUTPUT_DIR = ROOT / "outputs" / "sesi_01"
DOCS_DIR = ROOT / "docs" / "sesi_01"

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
DOCS_DIR.mkdir(parents=True, exist_ok=True)

print("PROJECT ROOT :", ROOT)
print("OUTPUT DIR   :", OUTPUT_DIR)
print("PYTHON       :", sys.executable)
```

## 1. Workspace Structure Check

Pastikan folder utama Workspace Starter tersedia dan peserta memahami fungsinya.

```python
workspace_rows = []

for rel in REQUIRED_DIRS:
    path = ROOT / rel
    workspace_rows.append({
        "path": rel,
        "exists": path.exists(),
        "is_directory": path.is_dir(),
    })

workspace_result = pd.DataFrame(workspace_rows)
workspace_structure_pass = bool(
    workspace_result["exists"].all() and
    workspace_result["is_directory"].all()
)

display(workspace_result)

print(
    "WORKSPACE STRUCTURE CHECK:",
    "PASS" if workspace_structure_pass else "FAIL"
)
```

## 2. Python Runtime Check

Notebook yang mengeksekusi cell ini membuktikan Python runtime aktif.

```python
python_result = pd.DataFrame([{
    "python_version": platform.python_version(),
    "implementation": platform.python_implementation(),
    "platform": platform.platform(),
    "executable": sys.executable,
}])

python_runtime_pass = True

display(python_result)
print("PYTHON RUNTIME CHECK: PASS")
```

## 3. pandas Smoke Test

Buktikan bahwa pandas dapat membuat DataFrame dan menjalankan agregasi sederhana.

```python
pandas_test_df = pd.DataFrame({
    "category": ["A", "A", "B"],
    "value": [10, 20, 30],
})

pandas_summary = (
    pandas_test_df
    .groupby("category", as_index=False)
    .agg(total_value=("value", "sum"))
)

expected = {"A": 30, "B": 30}
actual = dict(zip(
    pandas_summary["category"],
    pandas_summary["total_value"]
))

pandas_smoke_pass = actual == expected

display(pandas_test_df)
display(pandas_summary)

print(
    "PANDAS SMOKE TEST:",
    "PASS" if pandas_smoke_pass else "FAIL"
)
```

## 4. DuckDB Smoke Test

DuckDB adalah tool SQL utama bootcamp.

Smoke test:
1. import DuckDB;
2. register DataFrame pandas;
3. jalankan SQL;
4. cek hasil.

Jika belum tersedia, notebook tidak crash tetapi readiness tetap `NOT READY`.

```python
try:
    import duckdb
    DUCKDB_AVAILABLE = True
    duckdb_import_error = ""
except Exception as err:
    duckdb = None
    DUCKDB_AVAILABLE = False
    duckdb_import_error = f"{type(err).__name__}: {err}"

duckdb_smoke_pass = False
duckdb_version = None
duckdb_result = None

if DUCKDB_AVAILABLE:
    duckdb_version = getattr(duckdb, "__version__", "unknown")
    try:
        con = duckdb.connect()
        con.register("smoke_data", pandas_test_df)

        duckdb_result = con.sql(
            """
            SELECT
                category,
                SUM(value) AS total_value
            FROM smoke_data
            GROUP BY category
            ORDER BY category
            """
        ).df()

        duck_actual = dict(zip(
            duckdb_result["category"],
            duckdb_result["total_value"]
        ))
        duckdb_smoke_pass = duck_actual == {"A": 30, "B": 30}

        display(duckdb_result)
    except Exception as err:
        duckdb_import_error = f"{type(err).__name__}: {err}"
        duckdb_smoke_pass = False

print("DUCKDB AVAILABLE :", DUCKDB_AVAILABLE)
print("DUCKDB VERSION   :", duckdb_version)
print(
    "DUCKDB SMOKE TEST:",
    "PASS" if duckdb_smoke_pass else "NOT READY"
)

if duckdb_import_error:
    print("DETAIL:", duckdb_import_error)
```

## 5. Workspace File I/O Check

Uji `write → read → verify → delete` di `outputs/sesi_01/`.

Tidak ada raw file yang disentuh.

```python
io_test_path = OUTPUT_DIR / "_s1_write_read_test.txt"
test_payload = "S1_WORKSPACE_IO_OK"

file_io_pass = False
io_error = ""

try:
    io_test_path.write_text(test_payload, encoding="utf-8")
    read_back = io_test_path.read_text(encoding="utf-8")
    file_io_pass = read_back == test_payload
except Exception as err:
    io_error = f"{type(err).__name__}: {err}"
finally:
    if io_test_path.exists():
        io_test_path.unlink()

print(
    "WORKSPACE FILE I/O CHECK:",
    "PASS" if file_io_pass else "FAIL"
)

if io_error:
    print("DETAIL:", io_error)
```

## 6. Workspace Rules / Manifest Check

Workspace Starter final memiliki manifest di:

`docs/sesi_01/workspace_manifest.json`

```python
manifest_path = DOCS_DIR / "workspace_manifest.json"

manifest_pass = False
manifest_result = {}

if manifest_path.exists():
    try:
        manifest_result = json.loads(
            manifest_path.read_text(encoding="utf-8")
        )

        required_manifest_dirs = set(
            manifest_result.get("required_directories", [])
        )

        expected_manifest_dirs = {
            "data/raw",
            "data/processed",
            "notebooks/sesi_01",
            "sql/sesi_01",
            "outputs/sesi_01",
            "docs/sesi_01",
            "practice/sesi_01/source",
            "utils",
        }

        rules = manifest_result.get("rules", {})

        manifest_pass = (
            expected_manifest_dirs.issubset(required_manifest_dirs)
            and rules.get("raw_read_only_by_default") is True
            and rules.get("new_workspace_per_session") is False
            and rules.get("credentials_in_workspace") is False
        )
    except Exception as err:
        manifest_result = {
            "error": f"{type(err).__name__}: {err}"
        }

print("MANIFEST FOUND :", manifest_path.exists())
print(
    "WORKSPACE RULES CHECK:",
    "PASS" if manifest_pass else "NEEDS REVIEW"
)

if manifest_result:
    display(pd.DataFrame([manifest_result]))
```

## 7. Supporting Evidence Export

Supporting evidence ini **bukan output resmi tambahan**.

```python
checks = [
    {
        "check_id": "ENV-01",
        "check": "Workspace structure",
        "status": "PASS" if workspace_structure_pass else "FAIL",
        "detail": f"{int(workspace_result['exists'].sum())}/{len(REQUIRED_DIRS)} required paths found",
    },
    {
        "check_id": "ENV-02",
        "check": "Python runtime",
        "status": "PASS" if python_runtime_pass else "FAIL",
        "detail": platform.python_version(),
    },
    {
        "check_id": "ENV-03",
        "check": "pandas smoke test",
        "status": "PASS" if pandas_smoke_pass else "FAIL",
        "detail": f"pandas {pd.__version__}",
    },
    {
        "check_id": "ENV-04",
        "check": "DuckDB smoke test",
        "status": "PASS" if duckdb_smoke_pass else "NOT READY",
        "detail": (
            f"duckdb {duckdb_version}"
            if duckdb_smoke_pass
            else duckdb_import_error or "DuckDB smoke test belum PASS"
        ),
    },
    {
        "check_id": "ENV-05",
        "check": "Workspace file I/O",
        "status": "PASS" if file_io_pass else "FAIL",
        "detail": "temporary write/read/delete in outputs/sesi_01",
    },
    {
        "check_id": "ENV-06",
        "check": "Workspace rules manifest",
        "status": "PASS" if manifest_pass else "NEEDS REVIEW",
        "detail": str(manifest_path),
    },
]

readiness_table = pd.DataFrame(checks)
display(readiness_table)

summary_csv = OUTPUT_DIR / "s1_environment_readiness.csv"
summary_json = OUTPUT_DIR / "s1_environment_readiness.json"

readiness_table.to_csv(
    summary_csv,
    index=False,
    encoding="utf-8-sig"
)

summary_payload = {
    "generated_at": datetime.now().isoformat(timespec="seconds"),
    "workspace_root": str(ROOT),
    "python_version": platform.python_version(),
    "pandas_version": pd.__version__,
    "duckdb_available": DUCKDB_AVAILABLE,
    "duckdb_version": duckdb_version,
    "checks": checks,
}

summary_json.write_text(
    json.dumps(summary_payload, indent=2, ensure_ascii=False),
    encoding="utf-8"
)

evidence_export_pass = (
    summary_csv.exists() and
    summary_json.exists()
)

print("EVIDENCE CSV :", summary_csv)
print("EVIDENCE JSON:", summary_json)
print(
    "EVIDENCE EXPORT:",
    "PASS" if evidence_export_pass else "FAIL"
)
```

## 8. Final Readiness Gate

`FINAL READINESS GATE: PASS` berarti environment dasar S1 siap dipakai untuk sesi berikutnya.

```python
final_checks = {
    "Workspace structure": workspace_structure_pass,
    "Python runtime": python_runtime_pass,
    "pandas smoke test": pandas_smoke_pass,
    "DuckDB smoke test": duckdb_smoke_pass,
    "Workspace file I/O": file_io_pass,
    "Workspace rules manifest": manifest_pass,
    "Supporting evidence exported": evidence_export_pass,
}

final_readiness_gate_pass = all(final_checks.values())

final_gate_table = pd.DataFrame([
    {
        "check": check,
        "status": "PASS" if passed else "NOT READY",
    }
    for check, passed in final_checks.items()
])

display(final_gate_table)

print()

if final_readiness_gate_pass:
    print("FINAL READINESS GATE: PASS")
    print("S1 ENVIRONMENT & SMOKE TEST: COMPLETE")
else:
    print("FINAL READINESS GATE: NOT READY")

    if not workspace_structure_pass:
        print("NEXT STEP: perbaiki/extract ulang Workspace Starter.")
    elif not python_runtime_pass:
        print("NEXT STEP: perbaiki Python runtime/kernel.")
    elif not pandas_smoke_pass:
        print("NEXT STEP: perbaiki instalasi pandas.")
    elif not duckdb_smoke_pass:
        print(
            "NEXT STEP: install/aktifkan DuckDB pada environment kelas, "
            "lalu Restart Kernel → Run All."
        )
    elif not file_io_pass:
        print("NEXT STEP: cek permission folder workspace.")
    elif not manifest_pass:
        print("NEXT STEP: pastikan menggunakan Workspace Starter final S1.")
    elif not evidence_export_pass:
        print("NEXT STEP: cek permission outputs/sesi_01/.")
```
