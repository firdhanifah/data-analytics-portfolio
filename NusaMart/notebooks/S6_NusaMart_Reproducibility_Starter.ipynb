````python
# Menemukan root TERDEKAT. Tidak mencari source secara rekursif ke workspace lama.
def find_project_root(start=None):
    start = Path(start or Path.cwd()).expanduser().resolve()
    for folder in (start, *start.parents):
        if (folder / "practice" / "sesi_05" / "source").is_dir():
            return folder
    raise FileNotFoundError(
        "PROJECT ROOT belum ditemukan. Buka folder workspace dengan practice/sesi_05/source, "
        "atau periksa PROJECT_ROOT_OVERRIDE."
    )

def resolve_under_root(user_value, root, default_relative):
    value = Path(user_value).expanduser() if str(user_value).strip() else Path(default_relative)
    return (value if value.is_absolute() else root / value).resolve()

PROJECT_ROOT = find_project_root(PROJECT_ROOT_OVERRIDE.strip() or None)
if RUN_MODE != "MAIN" and (SOURCE_DIR_OVERRIDE.strip() or OUTPUT_DIR_OVERRIDE.strip()):
    raise ValueError("FRESH RUN: kosongkan SOURCE_DIR_OVERRIDE dan OUTPUT_DIR_OVERRIDE.")
SOURCE_DIR = resolve_under_root(SOURCE_DIR_OVERRIDE, PROJECT_ROOT, "practice/sesi_05/source")
OUTPUT_DIR = resolve_under_root(OUTPUT_DIR_OVERRIDE, PROJECT_ROOT, "outputs/sesi_06")
if not SOURCE_DIR.is_dir():
    raise FileNotFoundError(f"SOURCE DIR tidak tersedia: {SOURCE_DIR}")
if SOURCE_DIR == OUTPUT_DIR or SOURCE_DIR in OUTPUT_DIR.parents or OUTPUT_DIR in SOURCE_DIR.parents:
    raise ValueError("OUTPUT_DIR dan SOURCE_DIR harus terpisah dan tidak saling berada di dalam folder yang sama.")
if not OUTPUT_DIR.is_relative_to(PROJECT_ROOT):
    raise ValueError("Untuk keamanan, OUTPUT_DIR harus berada di dalam PROJECT_ROOT.")
if RUN_MODE != "MAIN":
    if not (PROJECT_ROOT / "clean_test_manifest.json").is_file():
        raise ValueError("FRESH RUN: clean_test_manifest.json tidak ada. Jalankan dari clean workspace Praktik 2B.")
    if SOURCE_DIR != (PROJECT_ROOT / "practice" / "sesi_05" / "source").resolve():
        raise ValueError("FRESH RUN harus menggunakan sumber di dalam clean workspace yang sama.")

# Persiapan: belum membuat/menulis output apa pun. Aktivasi run dilakukan pada Praktik 3/4/5.
OUTPUT_FILE = OUTPUT_DIR / "nusamart_analysis_ready.csv"
STATUS_FILE = OUTPUT_DIR / "s6_run_status.json"
VALIDATION_FILE = OUTPUT_DIR / "s6_validation_evidence.csv"
REPORT_FILE = OUTPUT_DIR / "s6_run_report.json"
# Helper checksum didefinisikan sebelum Praktik 2B (bukan baru saat LOAD).
def sha256_file(path):
    digest=hashlib.sha256()
    with Path(path).open("rb") as stream:
        for chunk in iter(lambda:stream.read(1024*1024),b""):
            digest.update(chunk)
    return digest.hexdigest()
````

### PRAKTIK 2A — Cari kebutuhan teknis yang belum ditulis

**Istilah baru:** *dependency* adalah **hal yang harus tersedia supaya proses bisa berjalan**. Misalnya file sumber, lokasi folder, atau package Python. *Path* artinya alamat/lokasi file. *Package* adalah perangkat lunak tambahan untuk membaca jenis data tertentu.

**Ikuti langkah ini:**
1. Pada Notebook, **klik ▶ Run pada cell kode pengecekan package tepat di bawah penjelasan ini**.
2. Baca tabel kolom **`package`** dan **`status`**. `PASS` = tersedia; `MISSING` = belum ada.
3. Jika ada `MISSING` atau error `ModuleNotFoundError`, **jangan lanjut**. Tunjukkan ke fasilitator agar package dipasang pada Python yang sedang digunakan.
4. Periksa hasil `SOURCE DIR` sebelumnya. Apakah file sumber masih menunjuk lokasi yang benar?
5. **Buka Working Paper → Sheet 02**, perbarui **C10/D10** (lokasi), **C11/D11** (urutan), dan **C12/D12** (syarat package/aturan).

**Contoh:** `pyarrow` adalah package pembaca Parquet. Jika `pyarrow` tidak tersedia, **jangan mengubah file Parquet**; minta bantuan pemasangan package.


````python
# Praktik 2A: apakah library pembaca XLSX / Parquet tersedia?
required_packages = {"pandas":"pandas", "numpy":"numpy", "pyarrow":"pyarrow", "openpyxl":"openpyxl"}
dependency_check = pd.DataFrame([
    {"package":label, "installed":bool(importlib.util.find_spec(import_name)),
     "status":"PASS" if importlib.util.find_spec(import_name) else "MISSING"}
    for label,import_name in required_packages.items()
])
display(dependency_check)
if not dependency_check["installed"].all():
    missing = dependency_check.loc[~dependency_check["installed"], "package"].tolist()
    print("DEPENDENCY CHECK: HOLD. Minta fasilitator membantu memasang:", ", ".join(missing))
    print("Contoh pada terminal environment yang tepat: python -m pip install", " ".join(missing))
    raise ModuleNotFoundError("Lanjut setelah dependency tersedia, lalu restart kernel dan Run All.")
print("DEPENDENCY CHECK: PASS")
````

### PRAKTIK 2B — Buat folder pengujian yang benar-benar baru

**Fresh workspace** = **folder baru** yang berisi enam input, Notebook, dan Master Prompt, **tanpa file hasil pengujian sebelumnya**. Kita membuatnya supaya tidak keliru memakai hasil kemarin.

Contoh isi folder yang akan dibuat otomatis:

```text
self_clean_01/
├── clean_test_manifest.json               ← daftar file yang disalin
├── practice/sesi_05/source/                ← enam input NusaMart
├── notebooks/S6_NusaMart_Reproducibility_Starter.ipynb
├── docs/sesi_06/S6_NusaMart_Master_Prompt.md
└── outputs/sesi_06/                        ← kosong saat pertama dibuat
```

**Ikuti langkah ini di Notebook awal (RUN_MODE masih MAIN):**
1. **Simpan Notebook** menggunakan **Ctrl+S**. Pastikan Notebook itu berada di folder `notebooks/` dalam proyek yang terbuka.
2. **Klik ▶ Run** pada cell kode panjang **`def prepare_clean_test_folder(...)`** tepat di bawah ini. Cell ini *memasang alat pembuat folder*, belum membuat folder.
3. **Pergi ke cell kode berikutnya** yang mempunyai tulisan `CREATE_FRESH_WORKSPACE = False`.
4. Pastikan `FRESH_FOLDER_NAME = "self_clean_01"`. Ubah **hanya** `CREATE_FRESH_WORKSPACE = True`, lalu **klik ▶ Run satu kali**.
5. Jika muncul `CLEAN TEST FOLDER READY` dan `INPUT COPIED: 6 | OLD OUTPUT COPIED: NO`, berarti folder berhasil dibuat. **Segera ubah kembali ke `False` dan simpan** (tidak perlu klik Run lagi).
6. **Buka folder `self_clean_01` sebagai folder proyek baru** melalui VS Code **File → Open Folder**, lalu buka Notebook dari `self_clean_01/notebooks/`.
7. **Jangan menjalankan pipeline dari Notebook awal lagi.** Praktik 3 dilakukan di Notebook yang ada dalam folder baru ini.

**Jika muncul `NOTEBOOK_FILE not found` atau folder sudah ada:** jangan menghapus hasil lama. Tunjukkan pesan error ke fasilitator; ia akan membantu memeriksa lokasi Notebook atau memilih nama folder baru.


````python
def prepare_clean_test_folder(destination, notebook_path, master_prompt_path=None, reference_run_reports=None):
    """Membuat fresh workspace baru; tidak pernah menghapus/overwrite folder yang ada.

    Dipakai untuk paket CSV FALLBACK S5. LIVE perlu akses Google Sheets yang dicatat di prompt.
    """
    destination=Path(destination).expanduser().resolve()
    notebook_path=Path(notebook_path).expanduser()
    if not notebook_path.is_absolute(): notebook_path=(PROJECT_ROOT/notebook_path).resolve()
    if destination.exists():
        raise FileExistsError(f"Folder tujuan sudah ada, pilih nama baru (tidak akan dihapus): {destination}")
    if destination==PROJECT_ROOT or destination in PROJECT_ROOT.parents:
        raise ValueError("Tidak boleh memakai project root/parent sebagai folder fresh.")
    if destination==SOURCE_DIR or destination in SOURCE_DIR.parents:
        raise ValueError("Folder fresh tidak boleh menimpa source S5.")
    if not notebook_path.is_file() or notebook_path.suffix.lower()!=".ipynb":
        raise FileNotFoundError(f"Salinan Notebook .ipynb belum ditemukan: {notebook_path}")
    if CHANNEL_SOURCE_MODE != "local_fallback":
        raise ValueError("Helper file lokal memakai CSV fallback. LIVE: siapkan akses source dan revisi instruksi secara terpisah.")
    keys=list(EXPECTED_LOCAL_FILES.keys())
    for key in keys:
        if not (SOURCE_DIR/EXPECTED_LOCAL_FILES[key]).is_file():
            raise FileNotFoundError(f"Input wajib belum tersedia: {EXPECTED_LOCAL_FILES[key]}")
    if master_prompt_path is not None:
        master_prompt_path=Path(master_prompt_path).expanduser()
        if not master_prompt_path.is_absolute():master_prompt_path=(PROJECT_ROOT/master_prompt_path).resolve()
        if not master_prompt_path.is_file():
            raise FileNotFoundError(f"Master Prompt belum ditemukan: {master_prompt_path}")
    target_source=destination/"practice"/"sesi_05"/"source"
    target_notebooks=destination/"notebooks"
    target_source.mkdir(parents=True,exist_ok=False)
    target_notebooks.mkdir(parents=True,exist_ok=False)
    (destination/"outputs"/"sesi_06").mkdir(parents=True,exist_ok=False)
    copied=[]
    for key in keys:
        src=SOURCE_DIR/EXPECTED_LOCAL_FILES[key]
        dst=target_source/src.name
        shutil.copy2(src,dst)
        if sha256_file(src)!=sha256_file(dst):
            raise IOError(f"Copy checksum tidak cocok: {src.name}")
        copied.append({"file":dst.name,"sha256":sha256_file(dst)})
    shutil.copy2(notebook_path,target_notebooks/"S6_NusaMart_Reproducibility_Starter.ipynb")
    if master_prompt_path is not None:
        docdir=destination/"docs"/"sesi_06";docdir.mkdir(parents=True,exist_ok=True)
        shutil.copy2(master_prompt_path,docdir/"S6_NusaMart_Master_Prompt.md")
    # Handoff referensi dilakukan secara eksplisit: tidak mengakses folder owner diam-diam.
    reference_run_reports = reference_run_reports or {}
    if not set(reference_run_reports).issubset({"SELF_CLEAN", "PEER_RUN"}):
        raise ValueError("Reference hanya boleh SELF_CLEAN atau PEER_RUN.")
    reference_copies = []
    for mode, supplied in reference_run_reports.items():
        supplied = Path(supplied).expanduser().resolve()
        if not supplied.is_file():
            raise FileNotFoundError(f"Report referensi {mode} tidak ditemukan: {supplied}")
        content = json.loads(supplied.read_text(encoding="utf-8"))
        if content.get("run_mode") != mode or content.get("export_status") != "PASS":
            raise ValueError(f"Report referensi {mode} bukan hasil run PASS yang sah.")
        if not content.get("source_hashes") or not content.get("output_content_sha256"):
            raise ValueError(f"Report referensi {mode} belum memiliki fingerprint lengkap.")
        folder = destination / "docs" / "sesi_06"
        folder.mkdir(parents=True, exist_ok=True)
        output = folder / f"reference_{mode.lower()}.json"
        shutil.copy2(supplied, output)
        reference_copies.append({"run_mode":mode,"filename":output.name,"report_sha256":sha256_file(output)})
    manifest={"created_at_utc":datetime.now(timezone.utc).isoformat(),
       "source_mode":"local_fallback", "source_files":copied,
       "notebook_copied":True, "master_prompt_copied":master_prompt_path is not None,
       "previous_outputs_copied":False, "reference_reports":reference_copies, "note":"Run notebook opened from fresh workspace, not source workspace"}
    (destination/"clean_test_manifest.json").write_text(json.dumps(manifest,indent=2),encoding="utf-8")
    print("CLEAN TEST FOLDER READY:",destination)
    print("INPUT COPIED:",len(copied),"| OLD OUTPUT COPIED: NO")
    print("REFERENCE REPORTS COPIED:", ", ".join(x["run_mode"] for x in reference_copies) or "NONE (SELF_CLEAN)")
    print("NEXT: buka",target_notebooks/"S6_NusaMart_Reproducibility_Starter.ipynb")
    return destination

# Praktik 2B CONTOH — tentukan path Notebook Anda, ganti nama folder untuk setiap run.
# CLEAN_TEST_ROOT = prepare_clean_test_folder(
#     PROJECT_ROOT / "practice" / "sesi_06" / "self_clean_01",
#     notebook_path=PROJECT_ROOT / "notebooks" / "S6_NusaMart_Reproducibility_Starter.ipynb",
#     master_prompt_path=MASTER_PROMPT_PATH,
 )
````

````python
def load_sqlite_table(db_path, preferred_table="payments"):
    with sqlite3.connect(db_path) as conn:
        names = pd.read_sql_query("SELECT name FROM sqlite_master WHERE type='table' ORDER BY name", conn)["name"].tolist()
        if preferred_table not in names:
            raise ValueError(f"SQLite memerlukan table '{preferred_table}', tersedia: {names}")
        return pd.read_sql_query('SELECT * FROM "payments"', conn)
orders = pd.read_csv(SOURCE_DIR / EXPECTED_LOCAL_FILES["orders"])
order_items = pd.read_parquet(SOURCE_DIR / EXPECTED_LOCAL_FILES["order_items"])
customers_safe = pd.read_excel(SOURCE_DIR / EXPECTED_LOCAL_FILES["customers_safe"])
payments = load_sqlite_table(SOURCE_DIR / EXPECTED_LOCAL_FILES["payments"])
products = pd.read_json(SOURCE_DIR / EXPECTED_LOCAL_FILES["products"])
if CHANNEL_SOURCE_MODE == "google_sheets":
    channel_campaign = pd.read_csv(GOOGLE_SHEETS_CSV_URL)
    channel_source_used = "Google Sheets LIVE (CSV export)"
else:
    channel_campaign = pd.read_csv(SOURCE_DIR / EXPECTED_LOCAL_FILES["channel_fallback"])
    channel_source_used = "LOCAL FALLBACK (CSV seed)"
load_run_id = RUN_ID
source_counts = {"orders":len(orders), "order_items":len(order_items), "customers_safe":len(customers_safe),
                 "payments":len(payments), "products":len(products), "channel_campaign":len(channel_campaign)}
source_summary = pd.DataFrame([
    {"source":name, "rows":size, "expected_S5":SOURCE_COUNT_S5[name],
     "status":"PASS" if size == SOURCE_COUNT_S5[name] else "REVIEW"}
    for name,size in source_counts.items()
])
display(source_summary)
print("CHANNEL SOURCE:", channel_source_used)
print("SOURCE LOAD:", "PASS" if source_summary["status"].eq("PASS").all() else "REVIEW — source berbeda dari paket S5")
````

### Langkah 4 — CHECK: schema, grain, dan key

**Untuk apa?** File dapat dibaca, tetapi mungkin tidak memiliki kolom atau key yang diperlukan. Memeriksa sebelum join mencegah kesalahan yang baru terlihat di akhir.

**LIHAT CONTOH:** `Orders.order_id` harus mengidentifikasi pesanan; `order_id` boleh berulang di Order Items karena satu order dapat memiliki banyak barang.

**COBA:** jalankan `schema_check` dan `key_check`. Perhatikan perbedaan antara key utama suatu source dan join key yang boleh berulang.

**KLIK RUN pada cell pemeriksaan di bawah.** `schema` = daftar kolom, `grain` = arti satu baris, dan `key` = kolom pengenal unik. Bila gagal, jangan memaksa lanjut JOIN.


````python
required_columns = {
 "Orders":{"order_id","customer_id","order_date","order_status","order_total"},
 "Order Items":{"order_item_id","order_id","product_id","quantity","line_total"},
 "Customers Safe":{"customer_id","city_group","customer_segment","join_date","customer_status"},
 "Payments":{"payment_id","order_id","amount_paid"},
 "Products":{"product_id","product_name","category","brand"},
 "Channel/Campaign":{"order_id","sales_channel","campaign_id","campaign_name"},
}
frames = {"Orders":orders,"Order Items":order_items,"Customers Safe":customers_safe,
          "Payments":payments,"Products":products,"Channel/Campaign":channel_campaign}
schema_check = pd.DataFrame([
    {"source":label, "missing_columns":", ".join(sorted(required_columns[label]-set(frame.columns))) or "-",
     "status":"PASS" if required_columns[label].issubset(set(frame.columns)) else "FAIL"}
    for label,frame in frames.items()
])
display(schema_check)
if not schema_check["status"].eq("PASS").all():
    raise ValueError("SCHEMA CHECK: HOLD — periksa missing_columns, jangan mengarang field.")

key_frames=[("Orders",orders,"order_id"),("Order Items",order_items,"order_item_id"),
            ("Customers Safe",customers_safe,"customer_id"),("Payments",payments,"payment_id"),
            ("Products",products,"product_id"),("Channel/Campaign",channel_campaign,"order_id")]
key_check=pd.DataFrame([
    {"source":label,"key":key,"rows":len(frame), "unique":int(frame[key].nunique(dropna=True)),
     "duplicate":int(frame[key].duplicated().sum()), "missing":int(frame[key].isna().sum()),
     "status":"PASS" if frame[key].is_unique and frame[key].notna().all() else "FAIL"}
    for label,frame,key in key_frames
])
display(key_check)
for label,frame,col in [("Orders",orders,"order_total"),
                        ("Items",order_items,"quantity"),("Items",order_items,"line_total"),
                        ("Payments",payments,"amount_paid")]:
    if not pd.api.types.is_numeric_dtype(frame[col]):
        raise TypeError(f"NON-NUMERIC SOURCE: {label}.{col}. Jangan konversi diam-diam.")
if not key_check["status"].eq("PASS").all():
    raise ValueError("KEY CHECK: HOLD — ditemukan key duplicate/null pada source.")
print("SCHEMA & KEY CHECK: PASS")
````

### Langkah 5 — TRANSFORM: memperkaya item dengan Products

**KLIK RUN pada cell kode berikut.** `TRANSFORM` artinya melengkapi data item dengan informasi Products. Hasilnya harus tetap satu baris per item, bukan bertambah karena join keliru.

### Langkah 6 — AGGREGATE: ringkas Items dan Payments per order

**Mengapa perlu?** Satu order bisa memiliki tiga barang dan dua pembayaran. Jika digabung mentah, masing-masing baris bisa terhitung berulang (fan trap). Kita membuat ringkasan per `order_id` sebelum final join.

**LIHAT CONTOH:** 3 baris item → satu baris ringkasan order; 2 transaksi pembayaran → satu total pembayaran order. Tidak berarti tiga item atau dua pembayaran dihapus.

**COBA:** jalankan 4B lalu 4C. Baca uniqueness serta kecocokan SUM sumber.

**KLIK RUN pada dua cell penggabungan ringkasan berikut secara urut.** `AGGREGATE` artinya merangkum banyak item/pembayaran menjadi total per order. Jangan menganggap jumlah item sama dengan jumlah unit barang.


````python
# 4B — NAMA FIELD mengikuti kontrak final S5 (bukan schema lama S6).
items_order=(items_enriched.groupby("order_id",as_index=False)
    .agg(item_count=("order_item_id","size"),
         quantity_total=("quantity","sum"),
         item_amount_total=("line_total","sum"),
         unique_product_count=("product_id","nunique"),
         unique_category_count=("category","nunique")))
print("ITEMS PER ORDER:",items_order.shape)
print("TOTAL SOURCE ITEM:",float(order_items["line_total"].sum()),
      "| AGG TOTAL:",float(items_order["item_amount_total"].sum()))
if (not items_order["order_id"].is_unique or
    not isclose(float(items_order["item_amount_total"].sum()),float(order_items["line_total"].sum()),abs_tol=.01)):
    raise ValueError("ITEM AGGREGATE: HOLD — grain/total belum sesuai.")
print("ITEM AGGREGATE: PASS")
````

**BACA OUTPUT:** `items_order` memiliki satu baris per order dan `item_amount_total` terjaga. Field hasil menggunakan nama S5 `item_count`, `quantity_total`, `item_amount_total`, `unique_product_count`, dan `unique_category_count`. Kelimanya tidak boleh diubah agar sekadar cocok dengan istilah lama pada WP S6.

````python
# 4C — dua pembayaran untuk satu order boleh; keduanya dijumlahkan.
payments_order=(payments.groupby("order_id",as_index=False)
    .agg(payment_count=("payment_id","size"),
         amount_paid_total=("amount_paid","sum")))
print("PAYMENTS PER ORDER:",payments_order.shape)
print("SOURCE PAYMENT:",float(payments["amount_paid"].sum()),
      "| AGG PAYMENT:",float(payments_order["amount_paid_total"].sum()))
if (not payments_order["order_id"].is_unique or
    not isclose(float(payments_order["amount_paid_total"].sum()),float(payments["amount_paid"].sum()),abs_tol=.01)):
    raise ValueError("PAYMENT AGGREGATE: HOLD — grain/total pembayaran belum sesuai.")
print("PAYMENT AGGREGATE: PASS")
````

**BACA OUTPUT:** `payments_order` satu baris per order dan `amount_paid_total` mempertahankan total source Payments. **JUDGMENT:** bila agregasi mengubah total secara tak wajar, STOP dan investigasi, bukan melanjutkan ke final join.

### Langkah 7 — JOIN: satukan enam sumber pada grain order

**Untuk apa?** Manajemen membutuhkan satu tabel utama dengan satu baris mewakili satu order untuk analisis berikutnya.

**COBA:** jalankan final join. Baca `FINAL SHAPE`, unique, duplicate dan nama kolom. Jangan langsung menyimpulkan benar hanya karena kode merge tidak error.

**KLIK RUN pada cell JOIN berikut.** `JOIN` berarti menggabungkan tabel berdasarkan pengenal yang sama (`order_id`/`customer_id`). Hasil akhirnya harus satu baris per order.


````python
final_integrated=(orders
 .merge(customers_safe[["customer_id","city_group","customer_segment","join_date","customer_status"]],
        on="customer_id",how="left",validate="many_to_one")
 .merge(items_order,on="order_id",how="left",validate="one_to_one")
 .merge(payments_order,on="order_id",how="left",validate="one_to_one")
 .merge(channel_campaign[["order_id","sales_channel","campaign_id","campaign_name"]],
        on="order_id",how="left",validate="one_to_one"))
missing_final = [c for c in S5_FINAL_COLUMNS if c not in final_integrated.columns]
extra_final = [c for c in final_integrated.columns if c not in S5_FINAL_COLUMNS]
if missing_final or extra_final:
    raise ValueError(f"FINAL SCHEMA: HOLD — missing {missing_final}, unexpected {extra_final}")
final_integrated=final_integrated[S5_FINAL_COLUMNS].copy()
final_build_run_id = RUN_ID
print("FINAL SHAPE:",final_integrated.shape)
print("UNIQUE order_id:",final_integrated["order_id"].nunique(),
      "| DUPLICATE:",final_integrated["order_id"].duplicated().sum())
print("COLUMN CONTRACT:", "PASS" if list(final_integrated.columns)==S5_FINAL_COLUMNS else "HOLD")
display(final_integrated.head(3))
````

````python
# Pemeriksaan kritis di luar 7 metrik WP (jangan hilangkan karena tidak ada di tabel minimum).
with sqlite3.connect(SOURCE_DIR/EXPECTED_LOCAL_FILES["payments"]) as conn:
    sqlite_tables = pd.read_sql_query("SELECT name FROM sqlite_master WHERE type='table'",conn)["name"].tolist()
    if "payments" not in sqlite_tables:
        raise ValueError("INDEPENDENT SQL CHECK: table payments tidak ditemukan")
    sqlite_payment_total=float(pd.read_sql_query(
        'SELECT SUM(amount_paid) AS payment_total FROM "payments"',conn).loc[0,"payment_total"])

source_counts_gate = all(source_counts[k] == SOURCE_COUNT_S5[k] for k in SOURCE_COUNT_S5)
source_unchanged_gate = all(
 sha256_file(SOURCE_DIR/EXPECTED_LOCAL_FILES[k]) == fingerprint
 for k,fingerprint in source_fingerprints.items())
source_integrity = [
 ("Source row counts S5", source_counts_gate, f"actual={source_counts}"),
 ("Raw file hashes unchanged", source_unchanged_gate, "SHA256 before vs after"),
 ("SUM(items) reconciles", isclose(float(final_integrated["item_amount_total"].sum()),
        float(order_items["line_total"].sum()),abs_tol=.01), "Source items vs final"),
 ("SUM(payment) vs SQLite independently", isclose(float(final_integrated["amount_paid_total"].sum()),
        sqlite_payment_total,abs_tol=.01),f"sqlite={sqlite_payment_total}"),
 ("Missing Customers = 0", int(final_integrated["customer_status"].isna().sum())==0,
        str(int(final_integrated["customer_status"].isna().sum()))),
 ("Missing Item Aggregate = 0", int(final_integrated["item_count"].isna().sum())==0,
        str(int(final_integrated["item_count"].isna().sum()))),
 ("Missing Payment Aggregate = 0", int(final_integrated["payment_count"].isna().sum())==0,
        str(int(final_integrated["payment_count"].isna().sum()))),
 ("Mapped channel source coverage", int(final_integrated["sales_channel"].notna().sum())==len(channel_campaign),
        f"mapped={int(final_integrated['sales_channel'].notna().sum())}, input_mapping={len(channel_campaign)}"),
 ("Campaign ID missing consistent", int(final_integrated["campaign_id"].isna().sum())==BASELINE["unmatched_channel_campaign"],
        str(int(final_integrated["campaign_id"].isna().sum()))),
 ("Campaign name missing consistent", int(final_integrated["campaign_name"].isna().sum())==BASELINE["unmatched_channel_campaign"],
        str(int(final_integrated["campaign_name"].isna().sum()))),
 ("Schema EXACTLY same as S5", list(final_integrated.columns)==S5_FINAL_COLUMNS,
        f"actual_columns={len(final_integrated.columns)}"),
]
integrity_summary=pd.DataFrame([
 {"check":n,"status":"PASS" if passed else "FAIL","evidence":detail}
 for n,passed,detail in source_integrity])
display(integrity_summary)

# HOTFIX P0 — pembuktian reproducibility membutuhkan pembandingan SELURUH ISI,
# bukan hanya ketujuh agregat. Fingerprint output dinormalisasi dengan urutan
# order_id dan urutan kolom kontrak S5; nilai null dijaga secara konsisten.
canonical_output_csv = final_integrated.sort_values("order_id", kind="mergesort")[S5_FINAL_COLUMNS].to_csv(
    index=False, lineterminator="\n", na_rep="<NA>", float_format="%.10g"
)
output_content_sha256 = hashlib.sha256(canonical_output_csv.encode("utf-8")).hexdigest()

REFERENCE_FILENAMES = {"SELF_CLEAN":"reference_self_clean.json", "PEER_RUN":"reference_peer_run.json"}
required_reference_modes = {
    "MAIN":[], "SELF_CLEAN":[], "PEER_RUN":["SELF_CLEAN"],
    "FINAL_RERUN":["SELF_CLEAN","PEER_RUN"],
}[RUN_MODE]
reference_details=[]
comparison_failures=[]
reference_ids=[]
for mode in required_reference_modes:
    reference_file=PROJECT_ROOT / "docs" / "sesi_06" / REFERENCE_FILENAMES[mode]
    if not reference_file.is_file():
        comparison_failures.append(f"Reference {mode} belum tersedia: {reference_file}")
        continue
    try:
        reference=json.loads(reference_file.read_text(encoding="utf-8"))
        if reference.get("run_mode")!=mode or reference.get("export_status")!="PASS":
            raise ValueError("Mode/status report tidak sah")
        if not isinstance(reference.get("run_id"), str) or not reference["run_id"]:
            raise ValueError("RUN_ID report tidak tersedia")
        if reference["run_id"] == RUN_ID or reference["run_id"] in reference_ids:
            raise ValueError("RUN_ID report sama/berulang")
        reference_ids.append(reference["run_id"])
        input_matches = reference.get("source_hashes") == source_fingerprints
        output_matches = reference.get("output_content_sha256") == output_content_sha256
        detail={"reference_mode":mode,"reference_run_id":reference["run_id"],
                "input_sha256_match":bool(input_matches),"output_content_sha256_match":bool(output_matches),
                "reference_report":str(reference_file),"reference_report_sha256":sha256_file(reference_file)}
        reference_details.append(detail)
        if not input_matches:
            comparison_failures.append(f"{mode}: isi/enam source berbeda (SHA256 INPUT)")
        if not output_matches:
            comparison_failures.append(f"{mode}: isi dataset berbeda meskipun agregat bisa sama (SHA256 OUTPUT)")
        if mode=="PEER_RUN" and reference.get("comparison_gate")!="PASS":
            comparison_failures.append("Referensi PEER_RUN belum terverifikasi terhadap SELF_CLEAN")
    except Exception as err:
        comparison_failures.append(f"Reference {mode} tidak valid: {err}")

# Manifest bersifat evidence setup; pastikan file referensi yang dipakai adalah
# yang disalin secara eksplisit melalui helper, bukan file ditanam belakangan.
if RUN_MODE in {"PEER_RUN","FINAL_RERUN"}:
    manifest=json.loads((PROJECT_ROOT / "clean_test_manifest.json").read_text(encoding="utf-8"))
    declared={x["run_mode"]:x for x in manifest.get("reference_reports",[])}
    for detail in reference_details:
        mode=detail["reference_mode"]
        if mode not in declared or declared[mode].get("report_sha256")!=detail["reference_report_sha256"]:
            comparison_failures.append(f"Reference {mode} tidak konsisten dengan manifest clean workspace")

comparison_gate = len(comparison_failures)==0
comparison_summary = pd.DataFrame(reference_details)
print("OUTPUT CONTENT SHA256:", output_content_sha256)
print("SOURCE SHA256 COUNT:",len(source_fingerprints))
if required_reference_modes:
    display(comparison_summary)
print("CROSS-RUN FINGERPRINT GATE:", "PASS" if comparison_gate else "HOLD",
      "(SELF_CLEAN menjadi BASELINE)" if not required_reference_modes else "")
if comparison_failures: print("FINGERPRINT ISSUES:", comparison_failures)

integrity_gate = bool(integrity_summary["status"].eq("PASS").all())
full_release_gate = bool(minimum_gate and reference_gate and integrity_gate and comparison_gate)
failures = (validation_summary.loc[validation_summary.status=="FAIL","check"].tolist()
            + integrity_summary.loc[integrity_summary.status=="FAIL","check"].tolist() + comparison_failures)
print("INTEGRITY:", "PASS" if integrity_gate else "HOLD")
print("PRE-EXPORT REVIEW GATE:", "PASS" if full_release_gate else "HOLD")
print("JUDGMENT: SELF_CLEAN PASS adalah baseline, bukan bukti peer reproducibility.")
if failures: print("HOLD / NEEDS REVIEW:", failures)
write_status("VALIDATED_PENDING_EXPORT" if full_release_gate else "HOLD",
             "Semua checks cocok, menunggu ekspor" if full_release_gate else "; ".join(failures))
````
