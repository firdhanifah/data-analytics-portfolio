## Aturan praktik: baca → isi/Run → periksa → lanjut

1. **Jangan klik `Run All`.** Ada langkah yang harus kamu isi **sebelum** hasil analisis terbuka. Jalankan cell kode satu per satu dengan tombol **▶ Run** di sebelah kirinya.
2. **Cell Markdown** (teks penjelasan) = baca. **Cell kode** bertanda `KLIK RUN` = jalankan, kecuali ada arahan khusus.
3. **Cell yang bertanya melalui `input()`** = klik ▶ Run, lalu **ketik jawaban di kotak pertanyaan yang muncul di bawah cell** dan tekan **Enter**. Jangan mengetik jawaban ke dalam kode.
4. Ada **empat tempat kamu mengisi**: (A) SPEC GATE: rumusan H1–H5, (B) INPUT VERIFIKASI: angka dari Excel, (C) FINAL JUDGMENT: keputusan dan alasan, (D) REFLEKSI: tiga jawaban.
5. Gunakan Antigravity untuk meninjau pekerjaan, **bukan mengganti hitungan manual** atau mengubah kriteria trainer.
6. **Kriteria dikunci sebelum melihat hasil.** Jangan ubah threshold atau sumber agar hasil cocok dengan dugaan.
7. Jika output tertulis `STOP`, `BELUM READY`, atau `NOT MATCH`, baca petunjuk di bawahnya dan minta bantuan trainer. Jangan terus menjalankan cell lain.
8. Semua contoh angka dalam visual dan teks penjelasan adalah **ilustrasi**, bukan jawaban dataset NusaMart.

### Kamus mini — istilah yang akan sering muncul

| Istilah | Arti sederhana |
|---|---|
| **Business Question** | Pertanyaan bisnis yang ingin dijawab menggunakan data. |
| **Hipotesis/klaim** | Dugaan yang akan diperiksa; bukan fakta sebelum diuji. |
| **Hipotesis dapat diuji (falsifiable)** | Dugaan yang memiliki kondisi jelas sehingga bukti bisa menunjukkan bahwa dugaan itu tidak didukung. |
| **Metrik (*metric*)** | Angka yang diukur, misalnya `order_total`. |
| **Dimensi (*dimension*)** | Cara membagi data, misalnya `sales_channel`, `customer_segment`, atau bulan. |
| **Ambang/batas kriteria** | Batas angka yang harus dilewati atau dipenuhi. Contoh: selisih harus **lebih dari 15%**. |
| **Efek/selisih relatif** | Seberapa besar perbedaan satu kelompok dibanding kelompok pembanding, dinyatakan dalam persen. |
| **Median** | Nilai tengah setelah angka diurutkan; biasanya tidak mudah berubah hanya karena ada beberapa nilai ekstrem. |
| **Mean/rata-rata** | Total nilai dibagi jumlah observasi. |
| **n / jumlah observasi** | Banyaknya baris pesanan yang masuk ke kelompok analisis. Dalam dataset ini, n biasanya berarti jumlah pesanan, bukan pelanggan unik. |
| **p-value** | Ukuran bukti statistik terhadap hipotesis uji tertentu. p-value kecil bukan berarti hipotesis bisnis pasti benar dan bukan ukuran besarnya efek. |
| **Mann–Whitney U** | Uji statistik untuk membandingkan dua kelompok berdasarkan peringkat nilai. Dalam latihan ini digunakan untuk contoh perbandingan satu arah; hasilnya tetap perlu dibaca bersama ukuran efek dan batasan data. |
| **Segmentasi** | Membandingkan hasil pada kelompok berbeda, misalnya kanal penjualan atau wilayah. |
| **Kontribusi** | Bagian dari total nilai yang tercatat. Kontribusi omzet bukan laba atau keuntungan. |
| **Robustness check** | Pemeriksaan tambahan untuk melihat apakah pola utama tetap masuk akal ketika dilihat dari sudut lain. |
| **Outlier/pencilan** | Nilai yang jauh berbeda dari sebagian besar data; perlu diperiksa, bukan otomatis dihapus. |
| **Cherry-picking** | Memilih hanya hasil yang mendukung dugaan awal dan mengabaikan hasil yang tidak cocok. |
| **Korelasi vs kausalitas** | Dua hal dapat terlihat berhubungan tanpa membuktikan bahwa salah satunya menyebabkan yang lain. |

## 4. Memahami lima kasus dalam *Valid Finding Challenge*

Kelima kasus menggunakan kolom yang tersedia pada dataset Sesi 6. Setiap kasus adalah **klaim latihan** yang harus diperiksa, bukan kesimpulan yang boleh langsung dipercaya.

### Apa maksud kalimat “kriteria latihan yang disediakan trainer”?

**Kriteria latihan** adalah aturan yang disepakati untuk kegiatan belajar ini agar semua peserta menilai klaim dengan standar yang sama. Contohnya, untuk H1 trainer menetapkan selisih median harus **lebih dari 15%**. Batas ini membantu kita berlatih memisahkan perbedaan yang cukup besar menurut aturan latihan dari perbedaan yang belum mencapai batas.

Angka tersebut **bukan hukum universal**. Tidak semua bisnis harus memakai 15%, 20%, `p < 0,05`, atau `n > 30`. Pada proyek nyata, kriteria ditentukan sesuai tujuan keputusan, risiko bisnis, kualitas data, dan rancangan analisis. Yang penting untuk latihan ini: kriteria ditulis **sebelum hasil utama dilihat** dan tidak diubah hanya karena hasilnya tidak sesuai dugaan.

### Cara membaca kriteria

- **`> 15%`** artinya harus lebih besar dari 15%; hasil tepat 15% belum memenuhi syarat.
- **`≥ 20%`** artinya 20% atau lebih sudah memenuhi bagian ambang efek tersebut.
- **`p < 0,05`** adalah batas statistik yang dipakai pada latihan, bukan probabilitas bahwa klaim benar.
- **`n > 30`** berarti lebih dari 30 observasi pada setiap kelompok. Ini batas praktis untuk kasus latihan, bukan jaminan sampel otomatis representatif atau analisis pasti valid.
- Jika satu kriteria wajib gagal, klaim tidak boleh disebut `SAH` hanya karena kriteria lain lolos.

Lihat penjelasan H1–H5 di bawah. Contoh angka pada setiap kasus adalah **ilustrasi buatan untuk belajar membaca aturan, bukan hasil aktual dari dataset NusaMart**.

**Petunjuk praktik:** H1–H5 sudah menjadi lima pertanyaan latihan. Kamu tidak diminta membuat lima pertanyaan baru. Bacalah metrik dan kondisi gugur masing-masing. **Saat SPEC GATE nanti**, kamu akan diminta menuliskan dengan kata-katamu sendiri melalui kotak input.

### H1 — Apakah median pesanan Store lebih tinggi daripada Marketplace?

**Jenis analisis:** segmentasi kanal penjualan.  
**Pertanyaan bisnis:** apakah nilai pesanan tipikal pada kanal Store cukup lebih tinggi dibanding Marketplace, atau perbedaannya terlalu kecil untuk mendukung klaim tersebut?

**Apa yang diukur?**
- `order_total`: nilai satu pesanan.
- `sales_channel`: kanal tempat pesanan dicatat.
- **Median**: nilai tengah pesanan pada masing-masing kanal.
- **Selisih median relatif**: persentase perbedaan median Store dibanding median Marketplace.

Rumus yang digunakan:

`(Median Store − Median Marketplace) ÷ Median Marketplace × 100%`

**Kriteria latihan yang harus diperiksa:**
1. selisih median relatif **lebih dari 15%**;
2. uji Mann–Whitney U satu arah memenuhi `p < 0,05` untuk arah `Store > Marketplace`;
3. jumlah pesanan `n` pada Store dan Marketplace masing-masing **lebih dari 30**; dan
4. saat hasil dipecah berdasarkan `customer_segment`, arah Store > Marketplace tetap terlihat pada setiap segmen yang dapat dibandingkan.

**Contoh ilustrasi (bukan hasil dataset):** median Store Rp120.000 dan median Marketplace Rp100.000. Selisihnya adalah `(120.000 − 100.000) ÷ 100.000 × 100% = 20%`. Angka 20% melewati ambang efek 15%, tetapi klaim **belum otomatis SAH**: kriteria statistik, ukuran sampel, dan pemeriksaan segmen masih harus diperiksa.

**Contoh kondisi gugur:** median Store Rp112.000 dan Marketplace Rp100.000 hanya berbeda 12%. Walaupun uji statistik menghasilkan p-value kecil, klaim tetap tidak memenuhi kriteria efek `> 15%`.

**Batasan dan kesalahan yang perlu dihindari:** mean dan median tidak sama; p-value bukan ukuran besarnya efek; dan perbedaan kanal tidak membuktikan sebab-akibat. Satu pelanggan juga dapat memiliki beberapa pesanan dan menggunakan lebih dari satu kanal. Karena itu, uji Mann–Whitney atas baris pesanan pada latihan ini adalah latihan membaca evidence; untuk kesimpulan inferensial formal, asumsi independensi dan kemungkinan pesanan berulang perlu ditangani secara khusus.

### H2 — Apakah median pesanan tidak pernah meningkat dari bulan ke bulan?

**Jenis analisis:** tren waktu.  
**Pertanyaan bisnis:** apakah median `order_total` selalu tetap atau turun dari Januari sampai Juni 2026, tanpa kenaikan pada satu pun perpindahan bulan?

**Apa yang diukur?**
- `order_date`: tanggal pesanan, yang dikelompokkan menjadi bulan.
- `order_total`: nilai pesanan.
- Median dihitung terpisah untuk Januari, Februari, Maret, April, Mei, dan Juni.
- Perubahan bulanan = median bulan sekarang dikurangi median bulan sebelumnya.

**Kriteria latihan:** semua enam bulan harus memiliki data dan setiap perubahan bulan-ke-bulan harus **≤ 0**. Dengan kata lain, tidak boleh ada satu pun kenaikan median.

**Contoh ilustrasi (bukan hasil dataset):** median bulanan adalah Rp150.000, Rp95.000, Rp100.000, Rp92.000, Rp90.000, dan Rp98.000. Ada kenaikan dari Februari ke Maret serta dari Mei ke Juni. Maka klaim “median tidak meningkat pada satu pun perpindahan bulan” gugur, walaupun median Juni masih lebih rendah daripada Januari.

**Mengapa aturannya ketat?** Klaimnya bukan “secara umum menurun”, tetapi “tidak meningkat dari bulan ke bulan”. Satu kenaikan saja cukup untuk menggugurkan kalimat yang sangat spesifik itu.

**Kesalahan yang perlu dihindari:** melihat hanya bulan pertama dan terakhir; mencampuradukkan median bulanan dengan total penjualan bulanan; atau menyebut tren ini membuktikan penyebab perubahan nilai pesanan.

### H3 — Apakah Store lebih tinggi daripada Marketplace di setiap wilayah yang cukup datanya?

**Jenis analisis:** segmentasi wilayah.  
**Pertanyaan bisnis:** apakah pola Store lebih tinggi daripada Marketplace tetap berlaku di semua `city_group` yang memiliki cukup pesanan untuk dibandingkan?

**Apa yang diukur?**
- `city_group`: kelompok wilayah.
- `sales_channel`: Store atau Marketplace.
- Median `order_total` pada masing-masing kanal di setiap wilayah.
- `n`: jumlah pesanan pada kanal tersebut di wilayah itu.

**Kriteria latihan:** suatu wilayah baru digunakan untuk menilai klaim jika memiliki setidaknya **10 pesanan Store dan 10 pesanan Marketplace**. Pada **semua** wilayah yang memenuhi batas tersebut, median Store harus lebih tinggi daripada median Marketplace. Wilayah yang tidak memenuhi batas tetap ditampilkan, tetapi tidak dipakai untuk memutuskan klaim.

**Contoh ilustrasi (bukan hasil dataset):**

| Wilayah | n Store | Median Store | n Marketplace | Median Marketplace | Arah |
|---|---:|---:|---:|---:|---|
| Wilayah A | 12 | Rp120.000 | 11 | Rp100.000 | Store lebih tinggi |
| Wilayah B | 14 | Rp105.000 | 13 | Rp115.000 | Marketplace lebih tinggi |

Kedua wilayah memenuhi batas jumlah pesanan. Karena di Wilayah B arah perbandingannya berlawanan, klaim “Store lebih tinggi di setiap wilayah yang memenuhi syarat” gugur.

**Kesalahan yang perlu dihindari:** hanya melaporkan wilayah yang mendukung dugaan awal; memasukkan wilayah dengan jumlah pesanan di bawah batas seolah-olah sama kuat; atau langsung menyebut perbedaan agregat dan per wilayah sebagai *Simpson's Paradox*. Pembalikan pola harus benar-benar diperiksa sebelum memakai istilah tersebut.

### H4 — Apakah median pesanan pelanggan Active minimal 20% lebih tinggi daripada Inactive?

**Jenis analisis:** perbandingan dua kelompok status pelanggan.  
**Pertanyaan bisnis:** apakah pesanan yang tercatat untuk pelanggan berstatus Active memiliki median `order_total` setidaknya 20% lebih tinggi dibanding pesanan pelanggan Inactive?

**Apa yang diukur?**
- `customer_status`: status Active atau Inactive pada data.
- `order_total`: nilai pesanan.
- Median dan selisih relatif median Active terhadap median Inactive.

**Kriteria latihan:** selisih median relatif **≥ 20%**, uji Mann–Whitney U satu arah memenuhi `p < 0,05`, dan jumlah pesanan pada kedua kelompok masing-masing **lebih dari 30**. Ketiga syarat harus diperiksa; memenuhi salah satunya saja tidak cukup.

**Contoh ilustrasi (bukan hasil dataset):** median Active Rp120.000 dan Inactive Rp100.000 menghasilkan selisih 20%. Bagian ambang efek memenuhi kriteria, tetapi jika p-value 0,08, kriteria statistik tidak terpenuhi sehingga klaim belum bisa dinyatakan sah menurut aturan latihan. Sebaliknya, selisih hanya 5% tidak memenuhi ambang efek walaupun p-value kecil.

**Catatan batasan penting:** dataset ini berisi baris pesanan, bukan satu baris per pelanggan. Seorang pelanggan dapat membuat beberapa pesanan, sehingga jumlah `n` di sini adalah jumlah pesanan dan bukan jumlah pelanggan unik. Gunakan hasil sebagai latihan membaca perbandingan tingkat pesanan; jangan mengubahnya menjadi klaim sebab-akibat atau kesimpulan formal tentang semua pelanggan tanpa rancangan analisis yang mempertimbangkan pesanan berulang.

**Kesalahan yang perlu dihindari:** menganggap status Active menyebabkan pesanan lebih besar, atau menganggap perbedaan median yang tampak sudah cukup tanpa memeriksa ambang, ukuran sampel, dan batasan unit analisis.

### H5 — Kampanye mana yang menyumbang total nilai pesanan terbesar?

**Jenis analisis:** kontribusi.  
**Pertanyaan bisnis:** di antara pesanan yang memiliki nama kampanye, kampanye mana yang mencatat jumlah `order_total` paling besar?

**Apa yang diukur?**
- `campaign_name`: nama kampanye yang tercatat pada pesanan.
- `order_total`: nilai pesanan.
- `sum(order_total)`: penjumlahan nilai pesanan untuk setiap nama kampanye.
- Peringkat: urutan kampanye dari jumlah nilai pesanan tertinggi ke terendah.

**Kriteria latihan:** `Free Shipping` harus berada di peringkat pertama di antara kampanye dengan `campaign_name` terisi. Pesanan dengan nama kampanye kosong harus dilaporkan terpisah, termasuk jumlah pesanan dan total `order_total`-nya.

**Contoh ilustrasi (bukan hasil dataset):** Free Shipping mencatat total Rp12.000.000, Payday Promo Rp10.000.000, dan Weekend Deals Rp9.000.000. Dalam contoh ini, Free Shipping menempati peringkat pertama sehingga klaim memenuhi kriteria peringkat. Jika Payday Promo ternyata Rp13.000.000, klaim gugur karena bukan Free Shipping yang tertinggi.

**Batas interpretasi:** total nilai pesanan adalah kontribusi nilai transaksi yang tercatat, **bukan laba atau keuntungan**. Peringkat tertinggi juga tidak membuktikan bahwa kampanye menyebabkan pesanan lebih besar, karena jumlah pesanan, jenis pelanggan, dan faktor lain dapat berbeda.

**Kesalahan yang perlu dihindari:** membandingkan total satu kampanye dengan rata-rata kampanye lain; mengabaikan pesanan tanpa nama kampanye; atau menulis “kampanye paling menguntungkan” ketika dataset tidak berisi biaya dan margin.

### PRAKTIK ANALYZE H1 — KLIK RUN, lalu baca hasilnya

**CASE:** apakah nilai pesanan Store lebih tinggi dibanding Marketplace? **KLIK RUN** pada cell H1 berikut, lalu baca `n`, `Median Store`, `Median Marketplace`, `Selisih median relatif`, dan `p-value`. Bandingkan dengan kriteria pada bagian SPECIFY. **Jangan tulis keputusan akhir dulu**, karena H1 masih harus diuji per segmen.


````python
# H1 — Segmentasi kanal: Store vs Marketplace
H1_READY = False
H1_CORE_PASS = False
h1_summary = None
h1_core_checks = None
h1_store = None
h1_marketplace = None
H1_EFFECT_PERCENT = None
H1_PVALUE = None
H1_SAMPLE_SUFFICIENT = False

if not ANALYSIS_ALLOWED:
    print("H1: STOP — selesaikan Input Gate dan Spec Gate terlebih dahulu.")
elif not DATA_READY:
    print("H1: STOP — dataset belum valid.")
else:
    h1_store = analysis_ready.loc[analysis_ready["sales_channel"].eq("Store"), "order_total"].dropna()
    h1_marketplace = analysis_ready.loc[analysis_ready["sales_channel"].eq("Marketplace"), "order_total"].dropna()
    h1_mean_store = float(h1_store.mean())
    h1_mean_marketplace = float(h1_marketplace.mean())
    h1_median_store = float(h1_store.median())
    h1_median_marketplace = float(h1_marketplace.median())
    H1_EFFECT_PERCENT = (h1_median_store - h1_median_marketplace) / h1_median_marketplace * 100 if h1_median_marketplace else float("nan")
    h1_test = mannwhitneyu(h1_store, h1_marketplace, alternative="greater", method="auto")
    H1_PVALUE = float(h1_test.pvalue)
    h1_summary = pd.DataFrame([
        {"Metrik": "Jumlah pesanan (n)", "Store": len(h1_store), "Marketplace": len(h1_marketplace)},
        {"Metrik": "Mean order_total (Rp)", "Store": h1_mean_store, "Marketplace": h1_mean_marketplace},
        {"Metrik": "Median order_total (Rp)", "Store": h1_median_store, "Marketplace": h1_median_marketplace},
    ])
    h1_core_checks = pd.DataFrame([
        {"Kriteria": "Selisih median relatif > 15%", "Aktual": round(H1_EFFECT_PERCENT, 2), "Syarat": "> 15%", "Lolos?": H1_EFFECT_PERCENT > 15},
        {"Kriteria": "n Store > 30", "Aktual": len(h1_store), "Syarat": "> 30", "Lolos?": len(h1_store) > 30},
        {"Kriteria": "n Marketplace > 30", "Aktual": len(h1_marketplace), "Syarat": "> 30", "Lolos?": len(h1_marketplace) > 30},
        {"Kriteria": "Mann–Whitney U satu arah p < 0,05", "Aktual": round(H1_PVALUE, 6), "Syarat": "p < 0,05", "Lolos?": H1_PVALUE < 0.05},
    ])
    H1_SAMPLE_SUFFICIENT = len(h1_store) > 30 and len(h1_marketplace) > 30
    H1_CORE_PASS = bool(h1_core_checks["Lolos?"].all())
    H1_READY = True
    print("H1 — HASIL ANALISIS (bukan keputusan akhir)")
    display(h1_summary)
    print(f"Selisih median relatif Store terhadap Marketplace: {H1_EFFECT_PERCENT:.2f}%")
    print(f"Mann–Whitney U satu arah: U = {float(h1_test.statistic):.0f}; p = {H1_PVALUE:.6f}")
    display(h1_core_checks)
    print("Kriteria inti:", "LOLOS" if H1_CORE_PASS else "TIDAK LOLOS")
````

### Cara membaca hasil H1

- **n** adalah jumlah pesanan pada setiap kanal, bukan jumlah pelanggan unik.
- **Mean** dapat lebih terpengaruh nilai ekstrem; bandingkan dengan median.
- Selisih relatif dihitung terhadap median Marketplace: `(median Store − median Marketplace) / median Marketplace × 100%`.
- Uji Mann–Whitney U di sini satu arah karena klaim secara spesifik menguji `Store > Marketplace`. Uji ini membandingkan peringkat/distribusi; interpretasi sebagai perbedaan median perlu kehati-hatian jika bentuk distribusi kelompok berbeda. p-value bukan ukuran besarnya efek dan tidak membuktikan sebab-akibat.
- Tabel kriteria inti belum mencakup robustness check dan pemeriksaan manual. Jangan menetapkan status akhir hanya dari tabel ini.

### PRAKTIK ANALYZE H2 — KLIK RUN

Cell H2 menghitung median Januari–Juni. Klik ▶ Run, lihat enam bulan dan kolom **Median naik?**. Jika ada bulan tanpa data, catat **bukti belum cukup**, bukan otomatis GUGUR.


````python
# H2 — Tren median bulanan Januari–Juni 2026
H2_READY = False
H2_PASS = False
h2_monthly = None
h2_monthly_changes = None
H2_INCREASE_COUNT = None
H2_MONTHS_COMPLETE = False

if not ANALYSIS_ALLOWED or not DATA_READY:
    print("H2: STOP — selesaikan Input Gate dan Spec Gate terlebih dahulu.")
else:
    h2_data = analysis_ready.loc[
        (analysis_ready["order_date"] >= pd.Timestamp("2026-01-01"))
        & (analysis_ready["order_date"] < pd.Timestamp("2026-07-01"))
    ].copy()
    h2_data["month"] = h2_data["order_date"].dt.to_period("M").astype(str)
    h2_monthly = h2_data.groupby("month")["order_total"].agg(n="count", median="median")
    expected_months = [str(p) for p in pd.period_range("2026-01", "2026-06", freq="M")]
    h2_monthly = h2_monthly.reindex(expected_months).reset_index().rename(columns={"month": "Bulan"})
    h2_monthly_changes = h2_monthly[["Bulan", "median"]].copy()
    h2_monthly_changes["Perubahan dari bulan sebelumnya (Rp)"] = h2_monthly_changes["median"].diff()
    h2_monthly_changes["Median naik?"] = h2_monthly_changes["Perubahan dari bulan sebelumnya (Rp)"].gt(0)
    H2_INCREASE_COUNT = int(h2_monthly_changes["Median naik?"].fillna(False).sum())
    complete_months = bool(h2_monthly["median"].notna().all() and h2_monthly["n"].notna().all())
    H2_MONTHS_COMPLETE = complete_months
    H2_PASS = bool(complete_months and H2_INCREASE_COUNT == 0)
    H2_READY = True
    print("H2 — MEDIAN ORDER_TOTAL PER BULAN")
    display(h2_monthly.rename(columns={"n": "Jumlah pesanan", "median": "Median order_total (Rp)"}))
    display(h2_monthly_changes.rename(columns={"median": "Median order_total (Rp)"}))
    print("Jumlah kenaikan median antarbulan:", H2_INCREASE_COUNT)
    print("Kecukupan bulan:", "LENGKAP" if H2_MONTHS_COMPLETE else "BELUM LENGKAP — TAHAN, jangan simpulkan GUGUR")
    print("Kriteria tren tidak meningkat:", "LOLOS" if H2_PASS else "TIDAK LOLOS / BELUM CUKUP BUKTI")
````

### Cara membaca hasil H2

Klaim H2 sangat spesifik: median tidak boleh naik pada satu pun transisi bulan-ke-bulan. Penurunan di awal periode tidak cukup jika bulan berikutnya kembali naik. `H2_INCREASE_COUNT` menghitung berapa kali median suatu bulan lebih tinggi daripada bulan sebelumnya.

````python
# H3 — Apakah Store memiliki median lebih tinggi di setiap city_group yang memenuhi syarat n?
H3_READY = False
H3_PASS = False
h3_city_table = None
h3_eligible_table = None
H3_EVIDENCE_SUFFICIENT = False

if not ANALYSIS_ALLOWED or not DATA_READY:
    print("H3: STOP — selesaikan Input Gate dan Spec Gate terlebih dahulu.")
else:
    h3_data = analysis_ready.loc[analysis_ready["sales_channel"].isin(["Store", "Marketplace"])].copy()
    h3_grouped = h3_data.groupby(["city_group", "sales_channel"])["order_total"].agg(n="count", median="median").reset_index()
    h3_city_rows = []
    for city in sorted(h3_data["city_group"].dropna().unique()):
        g = h3_grouped.loc[h3_grouped["city_group"].eq(city)].set_index("sales_channel")
        n_store = int(g.loc["Store", "n"]) if "Store" in g.index else 0
        n_market = int(g.loc["Marketplace", "n"]) if "Marketplace" in g.index else 0
        med_store = float(g.loc["Store", "median"]) if "Store" in g.index else np.nan
        med_market = float(g.loc["Marketplace", "median"]) if "Marketplace" in g.index else np.nan
        eligible = n_store >= 10 and n_market >= 10
        if n_store == 0 or n_market == 0:
            direction = "Tidak dapat dibandingkan"
        elif med_store > med_market:
            direction = "Store > Marketplace"
        elif med_store < med_market:
            direction = "Marketplace > Store"
        else:
            direction = "Median sama"
        h3_city_rows.append({
            "city_group": city,
            "n Store": n_store,
            "n Marketplace": n_market,
            "Median Store (Rp)": med_store,
            "Median Marketplace (Rp)": med_market,
            "Memenuhi syarat n ≥ 10 per kanal?": eligible,
            "Arah median": direction,
        })
    h3_city_table = pd.DataFrame(h3_city_rows, columns=["city_group", "n Store", "n Marketplace", "Median Store (Rp)", "Median Marketplace (Rp)", "Memenuhi syarat n ≥ 10 per kanal?", "Arah median"])
    h3_eligible_table = h3_city_table.loc[h3_city_table["Memenuhi syarat n ≥ 10 per kanal?"]].copy()
    H3_EVIDENCE_SUFFICIENT = len(h3_eligible_table) > 0
    H3_PASS = bool(H3_EVIDENCE_SUFFICIENT and h3_eligible_table["Arah median"].eq("Store > Marketplace").all())
    H3_READY = True
    print("H3 — HASIL PER CITY_GROUP. Kelompok dengan n kecil tetap ditampilkan untuk transparansi.")
    display(h3_city_table)
    print(f"Jumlah city_group yang memenuhi syarat n: {len(h3_eligible_table)}")
    print("Kecukupan segmen:", "ADA SEGMENT ELIGIBLE" if H3_EVIDENCE_SUFFICIENT else "TIDAK ADA — TAHAN, bukan otomatis GUGUR")
    print("Klaim Store > Marketplace pada semua kelompok yang memenuhi syarat:", "LOLOS" if H3_PASS else "TIDAK LOLOS / BELUM CUKUP BUKTI")
````

### Cara membaca hasil H3

## Robustness check wajib: pecah H1 berdasarkan `customer_segment`

Tujuannya adalah menguji apakah arah perbedaan agregat Store vs Marketplace tetap terlihat pada segmen pelanggan. Ini merupakan pemeriksaan ketahanan pola, bukan jaminan bahwa hasil berlaku untuk populasi lain.

````python
# ROBUSTNESS CHECK — customer_segment × sales_channel
ROBUSTNESS_READY = False
H1_ROBUSTNESS_PASS = False
H1_ROBUSTNESS_SUFFICIENT = False
H1_ROBUSTNESS_FAIL_EVIDENCED = False
h1_segment_table = None

if not ANALYSIS_ALLOWED or not H1_READY:
    print("ROBUSTNESS CHECK: STOP — jalankan H1 setelah Spec Gate READY terlebih dahulu.")
else:
    seg_data = analysis_ready.loc[analysis_ready["sales_channel"].isin(["Store", "Marketplace"])].copy()
    seg_grouped = seg_data.groupby(["customer_segment", "sales_channel"])["order_total"].agg(n="count", median="median").reset_index()
    seg_rows = []
    for segment in sorted(seg_data["customer_segment"].dropna().unique()):
        g = seg_grouped.loc[seg_grouped["customer_segment"].eq(segment)].set_index("sales_channel")
        if "Store" not in g.index or "Marketplace" not in g.index:
            seg_rows.append({"customer_segment": segment, "n Store": int(g.loc["Store", "n"]) if "Store" in g.index else 0,
                "n Marketplace": int(g.loc["Marketplace", "n"]) if "Marketplace" in g.index else 0,
                "Median Store (Rp)": float(g.loc["Store", "median"]) if "Store" in g.index else np.nan,
                "Median Marketplace (Rp)": float(g.loc["Marketplace", "median"]) if "Marketplace" in g.index else np.nan,
                "Arah": "Tidak dapat dibandingkan", "Catatan sampel": "BUKTI KURANG: kanal tidak lengkap"})
            continue
        n_store = int(g.loc["Store", "n"])
        n_market = int(g.loc["Marketplace", "n"])
        med_store = float(g.loc["Store", "median"])
        med_market = float(g.loc["Marketplace", "median"])
        if med_store > med_market:
            direction = "Store > Marketplace"
        elif med_store < med_market:
            direction = "Marketplace > Store"
        else:
            direction = "Median sama"
        seg_rows.append({
            "customer_segment": segment,
            "n Store": n_store,
            "n Marketplace": n_market,
            "Median Store (Rp)": med_store,
            "Median Marketplace (Rp)": med_market,
            "Arah": direction,
            "Catatan sampel": "Periksa n kecil" if min(n_store, n_market) < 10 else "n ≥ 10 per kanal",
        })
    h1_segment_table = pd.DataFrame(seg_rows, columns=["customer_segment", "n Store", "n Marketplace", "Median Store (Rp)", "Median Marketplace (Rp)", "Arah", "Catatan sampel"])
    eligible = h1_segment_table.loc[h1_segment_table["Catatan sampel"].eq("n ≥ 10 per kanal")].copy()
    H1_ROBUSTNESS_FAIL_EVIDENCED = bool(len(eligible) > 0 and eligible["Arah"].ne("Store > Marketplace").any())
    H1_ROBUSTNESS_SUFFICIENT = bool(len(h1_segment_table) > 0 and len(eligible) == len(h1_segment_table))
    H1_ROBUSTNESS_PASS = bool(H1_ROBUSTNESS_SUFFICIENT and not H1_ROBUSTNESS_FAIL_EVIDENCED)
    ROBUSTNESS_READY = True
    display(h1_segment_table)
    print("Pemeriksaan segmen:", "LENGKAP" if H1_ROBUSTNESS_SUFFICIENT else "BUKTI KURANG — ada segmen dengan n kecil/kanal hilang")
    print("Arah Store > Marketplace stabil di semua segmen layak:", "YA" if H1_ROBUSTNESS_PASS else "TIDAK / BELUM CUKUP BUKTI")
    print("Catatan: arah yang konsisten secara deskriptif tidak menghilangkan keterbatasan ukuran sampel tiap segmen.")
````

### Cara menyimpulkan robustness check

1. Bandingkan arah median pada setiap segmen yang memiliki kedua kanal.
2. Perhatikan ukuran sampel setiap sel; arah median pada kelompok kecil lebih tidak stabil.
3. Catat apakah arah berubah, besarnya efek berubah, atau bukti belum cukup.
4. Jangan menyebut *Simpson's Paradox* kecuali pola agregat dan pola subkelompok benar-benar memenuhi definisi pembalikan yang relevan.

### Pemeriksaan tambahan: sensitivitas terhadap outlier (opsional)

Outlier hanya **ditandai** menggunakan aturan IQR pada working copy. Kode berikut tidak mengubah `analysis_ready` atau file CSV sumber. Bandingkan mean/median sebelum dan setelah nilai yang ditandai dikeluarkan dari working copy; jangan menghapusnya hanya agar klaim lolos.

````python
# OPSIONAL — outlier sensitivity untuk Store dan Marketplace; source tidak diubah.
OUTLIER_CHECK_READY = False
outlier_sensitivity = None

if not ANALYSIS_ALLOWED or not H1_READY:
    print("Outlier check dilewati — selesaikan Spec Gate dan H1 terlebih dahulu.")
else:
    outlier_rows = []
    for channel in ["Store", "Marketplace"]:
        vals = analysis_ready.loc[analysis_ready["sales_channel"].eq(channel), "order_total"].dropna().astype(float)
        q1 = vals.quantile(0.25)
        q3 = vals.quantile(0.75)
        iqr = q3 - q1
        lower = q1 - 1.5 * iqr
        upper = q3 + 1.5 * iqr
        flagged = (vals < lower) | (vals > upper)
        retained = vals.loc[~flagged]
        outlier_rows.append({
            "sales_channel": channel,
            "n awal": int(vals.count()),
            "Outlier ditandai": int(flagged.sum()),
            "Mean awal (Rp)": float(vals.mean()),
            "Mean tanpa outlier (working copy) (Rp)": float(retained.mean()) if len(retained) else np.nan,
            "Median awal (Rp)": float(vals.median()),
            "Median tanpa outlier (working copy) (Rp)": float(retained.median()) if len(retained) else np.nan,
            "Batas bawah IQR": float(lower),
            "Batas atas IQR": float(upper),
        })
    outlier_sensitivity = pd.DataFrame(outlier_rows)
    display(outlier_sensitivity)
    OUTLIER_CHECK_READY = True
    print("Outlier ditandai untuk pemeriksaan; tidak dihapus dari source.")
````

**Checkpoint:** tiga klaim dipilih, minimal lima angka diperiksa, dan seluruh angka cocok sebelum kamu menerima status final. Jika belum selesai, keputusan klaim terkait harus **TAHAN**.


````python
# INPUT VERIFIKASI — INPUT PESERTA: KLIK RUN lalu ketik angka hasil Excel di output cell.
MANUAL_CASES_READY = False
MANUAL_SELECTED_CLAIMS = []
MANUAL_INPUTS = []
MANUAL_VERIFICATION_COMPLETE = False
MANUAL_VERIFICATION_PASS = False

verification_prompts = {
    "H1": [
        ("Median Store", "median order_total kanal Store (angka tanpa pemisah ribuan)"),
        ("Median Marketplace", "median order_total kanal Marketplace (angka tanpa pemisah ribuan)"),
        ("Selisih median relatif (%)", "selisih median relatif dalam persen, misalnya 22.17"),
    ],
    "H2": [
        ("Jumlah kenaikan median bulanan", "jumlah kenaikan median dari Jan ke Feb sampai Mei ke Jun (bilangan bulat)"),
    ],
    "H3": [
        ("Median Store Jabodetabek", "median Store pada city_group Jabodetabek"),
        ("Median Marketplace Jabodetabek", "median Marketplace pada city_group Jabodetabek"),
    ],
    "H4": [
        ("Median Active", "median order_total pelanggan Active"),
        ("Median Inactive", "median order_total pelanggan Inactive"),
        ("Selisih median relatif (%)", "selisih median relatif Active terhadap Inactive dalam persen"),
    ],
    "H5": [
        ("Total Free Shipping", "jumlah order_total untuk campaign_name Free Shipping"),
        ("Total Payday Promo", "jumlah order_total untuk campaign_name Payday Promo"),
    ],
}

def _read_numeric_value(prompt_text):
    while True:
        raw = input(prompt_text + " — masukkan angka tanpa pemisah ribuan: ").strip().replace(" ", "")
        try:
            return float(raw.replace(",", "."))
        except Exception:
            print("Format belum valid. Gunakan angka saja; untuk desimal boleh gunakan titik atau koma.")

if not ANALYSIS_ALLOWED or not DATA_READY:
    print("VERIFICATION INPUT: STOP — selesaikan Input Gate dan Spec Gate terlebih dahulu.")
else:
    while True:
        raw_selection = input("Pilih tepat 3 kasus untuk verifikasi Excel (untuk latihan terpandu: H1,H3,H5): ").strip().upper().replace(" ", "")
        selected = [x for x in raw_selection.split(",") if x]
        if len(selected) == 3 and len(set(selected)) == 3 and all(x in verification_prompts for x in selected):
            MANUAL_SELECTED_CLAIMS = selected
            break
        print("Pilihan belum valid. Pilih tepat tiga ID unik dari H1,H2,H3,H4,H5, dipisahkan koma.")

    for claim_id in MANUAL_SELECTED_CLAIMS:
        print(f"\n{claim_id}: masukkan hasil yang sudah kamu hitung di Excel.")
        for metric_name, metric_description in verification_prompts[claim_id]:
            entered = _read_numeric_value(metric_description)
            MANUAL_INPUTS.append({"Kasus": claim_id, "Metrik": metric_name, "Input Excel": entered})
    MANUAL_CASES_READY = len(MANUAL_SELECTED_CLAIMS) == 3
    print(f"Input diterima untuk {len(MANUAL_SELECTED_CLAIMS)} kasus dan {len(MANUAL_INPUTS)} angka. Jalankan cell Independent Recalculation berikutnya.")
````

````python
# INDEPENDENT RECALCULATION — membaca ulang CSV sumber, tidak memakai hasil object H1–H5.
manual_verification_table = None
MANUAL_VERIFICATION_COMPLETE = False
MANUAL_VERIFICATION_PASS = False

if not MANUAL_CASES_READY or DATA_PATH is None:
    print("MANUAL VERIFICATION: BELUM SIAP — pilih tiga kasus dan masukkan hasil Excel terlebih dahulu.")
else:
    try:
        verify_raw = pd.read_csv(DATA_PATH)
        verify_raw["order_total"] = pd.to_numeric(verify_raw["order_total"], errors="coerce")
        verify_raw["order_date"] = pd.to_datetime(verify_raw["order_date"], errors="coerce")
        expected_values = {}

        # H1: hitung ulang langsung dari CSV yang dibaca ulang.
        v_store = verify_raw.loc[verify_raw["sales_channel"].eq("Store"), "order_total"].dropna()
        v_market = verify_raw.loc[verify_raw["sales_channel"].eq("Marketplace"), "order_total"].dropna()
        v_h1_effect = (float(v_store.median()) - float(v_market.median())) / float(v_market.median()) * 100
        expected_values["H1"] = {
            "Median Store": float(v_store.median()),
            "Median Marketplace": float(v_market.median()),
            "Selisih median relatif (%)": round(v_h1_effect, 2),
        }

        # H2: bangun ulang tabel median bulanan dari source.
        v_h2 = verify_raw.loc[(verify_raw["order_date"] >= pd.Timestamp("2026-01-01")) & (verify_raw["order_date"] < pd.Timestamp("2026-07-01"))].copy()
        v_h2["month"] = v_h2["order_date"].dt.to_period("M").astype(str)
        v_monthly = v_h2.groupby("month")["order_total"].median().reindex([str(p) for p in pd.period_range("2026-01", "2026-06", freq="M")])
        v_increases = int(v_monthly.diff().gt(0).sum())
        expected_values["H2"] = {"Jumlah kenaikan median bulanan": v_increases}

        # H3: hitung ulang dua median Jabodetabek langsung dari source.
        v_city = verify_raw.loc[verify_raw["city_group"].eq("Jabodetabek")]
        expected_values["H3"] = {
            "Median Store Jabodetabek": float(v_city.loc[v_city["sales_channel"].eq("Store"), "order_total"].median()),
            "Median Marketplace Jabodetabek": float(v_city.loc[v_city["sales_channel"].eq("Marketplace"), "order_total"].median()),
        }

        # H4: hitung ulang median status pelanggan langsung dari source.
        v_active = verify_raw.loc[verify_raw["customer_status"].eq("Active"), "order_total"].dropna()
        v_inactive = verify_raw.loc[verify_raw["customer_status"].eq("Inactive"), "order_total"].dropna()
        v_h4_effect = (float(v_active.median()) - float(v_inactive.median())) / float(v_inactive.median()) * 100
        expected_values["H4"] = {
            "Median Active": float(v_active.median()),
            "Median Inactive": float(v_inactive.median()),
            "Selisih median relatif (%)": round(v_h4_effect, 2),
        }

        # H5: jumlah ulang dua kampanye langsung dari source.
        expected_values["H5"] = {
            "Total Free Shipping": float(verify_raw.loc[verify_raw["campaign_name"].eq("Free Shipping"), "order_total"].sum()),
            "Total Payday Promo": float(verify_raw.loc[verify_raw["campaign_name"].eq("Payday Promo"), "order_total"].sum()),
        }

        verification_rows = []
        for item in MANUAL_INPUTS:
            expected = float(expected_values[item["Kasus"]][item["Metrik"]])
            entered = float(item["Input Excel"])
            is_percent = "(%)" in item["Metrik"]
            is_count = "Jumlah kenaikan" in item["Metrik"]
            tolerance = 0.02 if is_percent else (0.0 if is_count else 0.5)
            matches = bool(np.isclose(entered, expected, atol=tolerance, rtol=0))
            verification_rows.append({
                "Kasus": item["Kasus"],
                "Metrik": item["Metrik"],
                "Input Excel": entered,
                "Hitung ulang independen": expected,
                "Status": "MATCH" if matches else "NOT MATCH",
            })
        manual_verification_table = pd.DataFrame(verification_rows)
        display(manual_verification_table)
        MANUAL_VERIFICATION_COMPLETE = len(MANUAL_SELECTED_CLAIMS) == 3 and len(manual_verification_table) >= 5
        MANUAL_VERIFICATION_PASS = bool(MANUAL_VERIFICATION_COMPLETE and manual_verification_table["Status"].eq("MATCH").all())
        print("Kasus yang diverifikasi:", ", ".join(MANUAL_SELECTED_CLAIMS))
        print("Jumlah angka diverifikasi:", len(manual_verification_table))
        print("MANUAL VERIFICATION:", "LULUS — semua angka MATCH" if MANUAL_VERIFICATION_PASS else "PERLU DICEK — ada NOT MATCH atau jumlah angka belum mencapai 5")
        if not MANUAL_VERIFICATION_PASS:
            print("Periksa formula, filter, tipe tanggal, dan pembulatan. Jangan mengubah source atau kriteria untuk membuat angka cocok.")
    except Exception as exc:
        print("MANUAL VERIFICATION: REVIEW — penghitungan ulang belum berhasil.")
        print("Jenis pesan:", type(exc).__name__)
        print("Pesan:", str(exc))
        print("Minta bantuan trainer/Agent untuk meninjau error; jangan menebak nilai.")
````

4. Jika `NOT MATCH`, buka lagi Excel dan periksa filter kanal/kota/kampanye, pemisah desimal, kolom angka, serta formula. Kemudian jalankan ulang `INPUT VERIFIKASI` dan `INDEPENDENT RECALCULATION` dengan angka **hasil Excel yang benar**.
5. **Jangan** mengubah dataset, threshold, atau menyalin angka Notebook ke Excel agar tampil MATCH. Jika belum selesai, status klaim terkait adalah **TAHAN**.

Sekarang tentukan keputusan untuk kelima klaim. Gunakan hasil analisis, pemeriksaan ketahanan, dan status verifikasi untuk kasus yang kamu pilih. Notebook akan menampilkan status yang didukung oleh aturan yang telah ditetapkan; kamu tetap harus meninjau evidence dan menuliskan alasanmu sendiri.

````python
# EVIDENCE SUMMARY — logika 3 status (bukti kurang tidak boleh otomatis GUGUR).
EVIDENCE_SUGGESTED = {key: "TAHAN" for key in ["H1", "H2", "H3", "H4", "H5"]}
EVIDENCE_REASON = {key: "Analisis atau pemeriksaan belum lengkap" for key in EVIDENCE_SUGGESTED}

if H1_READY and ROBUSTNESS_READY:
    if not H1_SAMPLE_SUFFICIENT:
        EVIDENCE_REASON["H1"] = "Jumlah pesanan satu/dua kanal di bawah syarat n>30"
    elif not H1_CORE_PASS:
        EVIDENCE_SUGGESTED["H1"] = "GUGUR"
        EVIDENCE_REASON["H1"] = "Efek/p-value kriteria inti gagal meskipun sampel memadai"
    elif H1_ROBUSTNESS_FAIL_EVIDENCED:
        EVIDENCE_SUGGESTED["H1"] = "GUGUR"
        EVIDENCE_REASON["H1"] = "Ada segmen dengan n memadai yang membalik/tidak mendukung arah efek"
    elif not H1_ROBUSTNESS_SUFFICIENT:
        EVIDENCE_REASON["H1"] = "Sebagian segmen terlalu kecil/tidak lengkap; perlu pemeriksaan lanjutan"
    else:
        EVIDENCE_SUGGESTED["H1"] = "SAH"
        EVIDENCE_REASON["H1"] = "Kriteria inti dan robustness segmen terpenuhi"

if H2_READY:
    if not H2_MONTHS_COMPLETE:
        EVIDENCE_REASON["H2"] = "Tidak semua bulan Jan–Jun memiliki data"
    else:
        EVIDENCE_SUGGESTED["H2"] = "SAH" if H2_PASS else "GUGUR"
        EVIDENCE_REASON["H2"] = "Jumlah kenaikan median bulan-ke-bulan: " + str(H2_INCREASE_COUNT)

if H3_READY:
    if not H3_EVIDENCE_SUFFICIENT:
        EVIDENCE_REASON["H3"] = "Tidak ada wilayah dengan n≥10 pada kedua kanal"
    else:
        EVIDENCE_SUGGESTED["H3"] = "SAH" if H3_PASS else "GUGUR"
        EVIDENCE_REASON["H3"] = "Ada wilayah memenuhi syarat; semua arah dicek tanpa memilih wilayah tertentu"

if H4_READY:
    if not H4_SAMPLE_SUFFICIENT:
        EVIDENCE_REASON["H4"] = "Jumlah pesanan Active/Inactive belum memenuhi n>30"
    else:
        EVIDENCE_SUGGESTED["H4"] = "SAH" if H4_PASS else "GUGUR"
        EVIDENCE_REASON["H4"] = "Efek median, p-value, dan ukuran sampel dibandingkan dengan kriteria"

if H5_READY:
    if not H5_EVIDENCE_SUFFICIENT:
        EVIDENCE_REASON["H5"] = "Tidak ada nilai kampanye bernama yang cukup untuk diperingkat"
    else:
        EVIDENCE_SUGGESTED["H5"] = "SAH" if H5_PASS else "GUGUR"
        EVIDENCE_REASON["H5"] = "Peringkat nilai transaksi kampanye bernama; pesanan tanpa kampanye dilaporkan"

# Klaim yang dipilih untuk verifikasi Excel tidak boleh dinyatakan SAH/GUGUR jika belum MATCH.
if MANUAL_SELECTED_CLAIMS:
    for claim_id in MANUAL_SELECTED_CLAIMS:
        if not MANUAL_VERIFICATION_COMPLETE or manual_verification_table is None:
            EVIDENCE_SUGGESTED[claim_id] = "TAHAN"
            EVIDENCE_REASON[claim_id] = "Verifikasi Excel untuk klaim pilihan belum selesai"
        else:
            claim_rows = manual_verification_table.loc[manual_verification_table["Kasus"].eq(claim_id)]
            if claim_rows.empty or not claim_rows["Status"].eq("MATCH").all():
                EVIDENCE_SUGGESTED[claim_id] = "TAHAN"
                EVIDENCE_REASON[claim_id] = "Hasil hitung ulang Excel masih NOT MATCH"

if PANDAS_READY:
    evidence_rows = [
        {"Kasus": case, "Status didukung evidence": EVIDENCE_SUGGESTED[case], "Alasan ringkas": EVIDENCE_REASON[case]}
        for case in ["H1", "H2", "H3", "H4", "H5"]
    ]
    display(pd.DataFrame(evidence_rows))
    print("Baca alasan tiap status. TAHAN berarti belum cukup bukti, bukan otomatis GUGUR.")
````

## JUDGE — Di mana kamu mengetik keputusan H1–H5?

**KAMU ISI DI SINI:** setelah menjalankan **EVIDENCE SUMMARY**, klik **▶ Run** pada cell `INPUT PESERTA — FINAL JUDGMENT` tepat di bawah.

Notebook akan bertanya untuk setiap H:

1. **Keputusan H1:** ketik `SAH`, `GUGUR`, atau `TAHAN`, lalu Enter.
2. **Alasan H1:** tulis **angka/metrik, ambang kriteria, dan apakah verifikasi/segmentasi mendukung**, lalu Enter.
3. Ulangi untuk **H2, H3, H4, H5**.

Contoh alasan yang memenuhi standar **(hanya ilustrasi)**: "H1 GUGUR karena efek median 8% di bawah batas >15%, walaupun n tiap kelompok mencukupi." Jangan menyalin contoh sebagai jawaban dataset.

**BACA OUTPUT:** setelah 10 jawaban, periksa tabel `Pemeriksaan`. Jika ada `PERIKSA LAGI`, tinjau bukti dan alasan. FINAL CHECK tidak akan LOLOS selama ada keputusan yang tidak konsisten dengan evidence.


````python
# INPUT PESERTA — final judgment dan alasan berbasis evidence.
allowed_judgments = {"SAH", "GUGUR", "TAHAN"}
FINAL_JUDGMENTS = {}

if not all([H1_READY, H2_READY, H3_READY, H4_READY, H5_READY, ROBUSTNESS_READY]):
    print("FINAL JUDGMENT: BELUM SIAP — pastikan H1–H5 dan robustness check sudah dijalankan.")
else:
    case_labels = {
        "H1": "H1 — Store vs Marketplace",
        "H2": "H2 — Tren median bulanan",
        "H3": "H3 — Konsistensi antarkota",
        "H4": "H4 — Active vs Inactive",
        "H5": "H5 — Kontribusi kampanye",
    }
    for claim_id, label in case_labels.items():
        while True:
            status = input(f"Keputusan akhir {label} (SAH/GUGUR/TAHAN): ").strip().upper()
            if status in allowed_judgments:
                break
            print("Pilihan belum valid. Ketik SAH, GUGUR, atau TAHAN.")
        while True:
            reason = input(f"Alasan berbasis angka/kriteria untuk {claim_id} (wajib diisi): ").strip()
            if reason:
                break
            print("Alasan tidak boleh kosong. Sebutkan evidence dan kriteria yang relevan.")
        FINAL_JUDGMENTS[claim_id] = {"Keputusan peserta": status, "Alasan": reason}

    final_rows = []
    for claim_id, label in case_labels.items():
        final_rows.append({
            "Kasus": label,
            "Keputusan peserta": FINAL_JUDGMENTS[claim_id]["Keputusan peserta"],
            "Status didukung evidence": EVIDENCE_SUGGESTED[claim_id],
            "Pemeriksaan": "KONSISTEN" if FINAL_JUDGMENTS[claim_id]["Keputusan peserta"] == EVIDENCE_SUGGESTED[claim_id] else "PERIKSA LAGI",
            "Alasan peserta": FINAL_JUDGMENTS[claim_id]["Alasan"],
        })
    final_judgment_table = pd.DataFrame(final_rows)
    display(final_judgment_table)
    print("Jika ada PERIKSA LAGI, tinjau ulang evidence dan alasan. Jangan mengubah threshold agar keputusan tampak konsisten.")
````
