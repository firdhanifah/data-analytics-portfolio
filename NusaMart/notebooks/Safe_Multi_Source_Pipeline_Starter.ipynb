## TAHAP 3 — Grain, Candidate Key & Relationship

**CASE — Situasi SwiftBite:** Items dan Payments boleh mengulang `order_id`, tetapi `order_item_id` dan `payment_id` seharusnya unik.

**TUJUAN:** Menguji key sesuai unit observasi sebelum join.

**PAHAMI ISTILAH:** grain = arti satu row; candidate key = calon pengenal unik; foreign key = rujukan key antar-source; cardinality = jumlah pasangan antar-record.

**WORKING PAPER BARU:** Lengkapi `01_Input_dan_Safe_Data!C13:D18` dari `input_inventory_df` dan `grain_key_review`. `order_id` yang berulang pada Items/Payments sesuai grain dan bukan otomatis error.

> Ikuti cell dari atas ke bawah. Jalankan code cell dengan **Shift + Enter** atau tombol ▶ Run. *Markdown cell* adalah penjelasan yang tidak perlu dijalankan.

**LIHAT CONTOH:** Dua pembayaran pada satu order adalah dua attempts, bukan duplikasi. Jangan menghapus record hanya karena `order_id` berulang di Payments.

```python
# ▶ WAJIB JALANKAN — kode utama siap pakai; jangan ubah fungsi.
def key_check(df,key_cols,dataset_name):
    cols=[key_cols] if isinstance(key_cols,str) else list(key_cols)
    missing_cols=[c for c in cols if c not in df.columns]
    if missing_cols: return {'dataset':dataset_name,'rows':len(df),'unique_key_count':0,'missing_key_rows':len(df),
                             'duplicate_key_rows':len(df),'candidate_key_unique':False,'error':'kolom hilang: '+str(missing_cols)}
    miss=df[cols].isna().any(axis=1)
    dup=df.duplicated(subset=cols,keep=False)
    return {'dataset':dataset_name,'rows':len(df),'unique_key_count':df[cols].drop_duplicates().shape[0],
            'missing_key_rows':int(miss.sum()),'duplicate_key_rows':int(dup.sum()),
            'candidate_key_unique':bool(not miss.any() and not dup.any()),'error':''}

grain_key_review = pd.DataFrame([key_check(df,SOURCE_KEYS[name],name) for name,df in pm2_sources.items()])
if len(grain_key_review):
    display(grain_key_review)
    print('KEY CHECK:', 'PASS' if bool(grain_key_review.candidate_key_unique.all()) else 'FAIL — periksa dataset pada kolom dataset')
else: print('KEY CHECK: PENDING (input belum terbaca)')
```

**BACA OUTPUT → INTERPRETASI → JUDGMENT → ISI WORKING PAPER**

`duplicate_key_rows=0` dan `missing_key_rows=0` berarti candidate key sesuai harapan. Kriteria sumber Items/Payments menggunakan `order_item_id`/`payment_id`, **bukan** `order_id`.

**PENJELASAN OUTPUT:** Pada `grain_key_review`, nilai `candidate_key_unique=True` dengan `missing_key_rows=0` dan `duplicate_key_rows=0` berarti key sumber valid. Repeated `order_id` pada Payments/Items tidak diuji sebagai key tunggal.

## TAHAP 5 — Pseudonymization & Safe Relationship

**CASE — Situasi SwiftBite:** Menghapus `customer_id` tanpa pengganti merusak hubungan Orders–Customers.

**TUJUAN:** Membentuk customer_key deterministik, menjaga join dan menghilangkan PII dari safe copy.

**PAHAMI ISTILAH:** deterministic = sama hasilnya jika input dan parameter sama; salt = konfigurasi rahasia hashing; pseudonymization = mengganti ID asli dengan key pengganti.

**WORKING PAPER BARU:** Lengkapi `01_Input_dan_Safe_Data!F24:H30` (F bukti dari `safe_check_df`, G status implementasi, H catatan). Evidence V-01 sampai V-04 *hanya* dicatat pada sheet 03 setelah Tahap 11.

> Ikuti cell dari atas ke bawah. Jalankan code cell dengan **Shift + Enter** atau tombol ▶ Run. *Markdown cell* adalah penjelasan yang tidak perlu dijalankan.

**LIHAT CONTOH:** `C0001` pada Customers dan `C0001` pada Orders harus menghasilkan **customer_key yang sama**. Hash aktual tidak perlu dibaca atau ditulis lengkap; cukup bukti konsistensinya.

**COBA:** jalankan definisi helper. Kamu tidak perlu menulis algoritma SHA-256 dari nol. `VARIANT_ID` menjadi namespace sehingga ID yang sama dari varian berbeda tidak otomatis menghasilkan key yang sama. Jika salt PENDING, kembali ke Tahap 1 dan isi **prompt tertutup**; jangan mengetik secret pada code cell atau mengirimkannya ke mentor.

```python
# ▶ WAJIB JALANKAN — kode utama siap pakai; jangan ubah fungsi.
def deterministic_hash(value,salt,variant_id):
    if pd.isna(value): return pd.NA
    if not salt: raise ValueError('PENGAMAN DATA BELUM SIAP — jalankan ulang Tahap 1 terlebih dahulu')
    payload=json.dumps([str(variant_id),str(value),str(salt)],ensure_ascii=False,separators=(',',':')).encode('utf-8')
    return hashlib.sha256(payload).hexdigest()

def build_safe_customer_data(customers_input,orders_input,salt,variant_id):
    customers=customers_input.copy(deep=True); orders=orders_input.copy(deep=True)
    for name,df in [('Customers',customers),('Orders',orders)]:
        if 'customer_id' not in df.columns: raise ValueError(name+': customer_id belum ada')
        if df.customer_id.isna().any(): raise ValueError(name+': customer_id null; cek PM1 handoff')
    customers['customer_key']=customers['customer_id'].map(lambda x:deterministic_hash(x,salt,variant_id))
    orders['customer_key']=orders['customer_id'].map(lambda x:deterministic_hash(x,salt,variant_id))
    customers=customers.drop(columns=[c for c in DIRECT_PII_FIELDS | RAW_IDENTIFIER_FIELDS if c in customers.columns])
    orders=orders.drop(columns=['customer_id'])
    return customers,orders

def safe_schema_check(df):
    return sorted(set(df.columns)&(DIRECT_PII_FIELDS|RAW_IDENTIFIER_FIELDS))

def fk_unmatched_count(child,parent,child_key,parent_key):
    if child_key not in child or parent_key not in parent: return len(child)
    return int((~child[child_key].isin(set(parent[parent_key].dropna()))).sum())

RUN_READY=bool(INPUT_FILES_READY and IDENTITY_READY and CONTROLLED_SALT)
customers_safe,orders_safe=(build_safe_customer_data(pm2_sources['Customers Clean'],pm2_sources['Orders Clean'],CONTROLLED_SALT,VARIANT_ID)
    if RUN_READY else (None,None))
if RUN_READY:
    safe_check_df=pd.DataFrame([{'check':'forbidden customer fields','actual':len(safe_schema_check(customers_safe))},
       {'check':'forbidden order fields','actual':len(safe_schema_check(orders_safe))},
       {'check':'unmatched customer_key','actual':fk_unmatched_count(orders_safe,customers_safe,'customer_key','customer_key')},
       {'check':'duplicate customer_key','actual':int(customers_safe.customer_key.duplicated().sum())}])
    display(safe_check_df)
    print('SAFE TRANSFORMATION:', 'PASS' if (safe_check_df.actual==0).all() else 'FAIL')
else: print('SAFE TRANSFORMATION: PENDING — perlu input lengkap, identitas dan SWIFTBITE_PM2_SALT')
```

**BACA OUTPUT → INTERPRETASI → JUDGMENT → ISI WORKING PAPER**

Semua metric dalam `safe_check_df` seharusnya 0. Output yang ditampilkan tidak memuat isi PII. PS-03 (pseudonymization rerun) dilengkapi nanti pada Tahap 12; SR-04 final schema setelah Tahap 10.

**ISI WORKING PAPER BARU:** Lengkapi `01_Input_dan_Safe_Data!F24:H30` (F bukti dari `safe_check_df`, G status implementasi, H catatan). Evidence V-01 sampai V-04 *hanya* dicatat pada sheet 03 setelah Tahap 11.



**PENJELASAN OUTPUT:** `safe_check_df` harus menunjukkan 0 forbidden fields, 0 unmatched customer key, dan 0 duplicate customer_key. Jika salt kosong, status PENDING dan jangan lanjut hashing.

**CHECKPOINT:** Jika data atau evidence belum sesuai, jangan menandai PASS di Working Paper. Perbaiki tahap yang disebutkan output dan jalankan ulang pemeriksaan.

## TAHAP 8 — Payment Attempts dan Monetary Aggregation

**CASE — Situasi SwiftBite:** Pembayaran dapat gagal lalu berhasil, atau berhasil kemudian di-refund. Mengambil satu payment record secara sembarangan dapat mengubah net payment.

**TUJUAN:** Menghitung semua attempts dan menentukan status/method secara deterministic.

**PAHAMI ISTILAH:** attempt = satu catatan percobaan/event pembayaran; refund = pengembalian uang; tie-breaker = aturan urutan jika nilai sebelumnya sama.

**WORKING PAPER BARU:** Isi `02_Integrasi_dan_Rerun!E13:H13` (E = `len(payment_agg)`, F = status, G = `payment_agg`, H = catatan). V-08 dan V-09 nanti di sheet 03 baris 19–20.

> Ikuti cell dari atas ke bawah. Jalankan code cell dengan **Shift + Enter** atau tombol ▶ Run. *Markdown cell* adalah penjelasan yang tidak perlu dijalankan.

**LIHAT CONTOH:** O001 failed Rp50.000 lalu successful Rp50.000 → payment_attempt_count 2, successful 50.000, refunded 0, net 50.000. Untuk successful → refunded penuh Rp50.000, net 0.

**Aturan resmi:** urutkan `attempt_number`, kemudian `payment_timestamp`, kemudian `payment_id`. Status final = record terakhir. Metode final = metode successful terakhir jika ada, jika tidak metode record terakhir.

**LATIHAN OPSIONAL (boleh dilewati):** kode berikut menguji rumus; pastikan `net = sukses - refund`. Failed tidak ditambah ke successful.

```python
# ★ LATIHAN OPSIONAL — boleh dilewati; bukan syarat penilaian.
contoh_successful=50000
contoh_refunded=50000
contoh_net=contoh_successful-contoh_refunded
auto_check_contoh = contoh_net==0
print('Net payment setelah refund penuh:',contoh_net)
print('CHECK CONTOH:', 'PASS' if auto_check_contoh else 'FAIL — net harus successful dikurangi refunded')
```

```python
# ★ LATIHAN OPSIONAL — boleh dilewati; bukan syarat penilaian.
# ★ LATIHAN OPSIONAL — cukup baca hasil; jangan ubah status bila belum memahami Python.
# Fokus: failed tetap ada sebagai attempt, tetapi tidak ditambah ke uang diterima.
STATUS_BERHASIL = 'successful'  # Ubah hanya saat latihan; kembalikan ke successful
if INPUT_FILES_READY:
    payments_latihan = pm2_sources['Payments Clean']
    uang_sukses = payments_latihan.loc[payments_latihan.payment_status.eq(STATUS_BERHASIL),'payment_amount'].sum()
    uang_refund = payments_latihan.loc[payments_latihan.payment_status.eq('refunded'),'payment_amount'].sum()
    print('Successful amount:',int(uang_sukses),'| Refunded amount:',int(uang_refund))
    print('Net amount:',int(uang_sukses-uang_refund))
else: print('PENDING — Payments Clean belum terbaca')
```

```python
# ▶ WAJIB JALANKAN — kode utama siap pakai; jangan ubah fungsi.
PAYMENT_STATUSES={'failed','successful','refunded'}
def build_payment_aggregate(payments_df):
    df=payments_df.copy(deep=True)
    required={'order_id','payment_id','payment_status','payment_method','payment_amount','payment_timestamp','attempt_number'}
    if required-set(df): raise ValueError('PAYMENT COLUMNS MISSING: '+str(required-set(df)))
    if df['payment_id'].isna().any() or df['payment_id'].duplicated().any() or df['order_id'].isna().any():
        raise ValueError('PAYMENT KEY FAIL — payment_id duplikat/null atau order_id null')
    if not set(df.payment_status.dropna().unique()).issubset(PAYMENT_STATUSES) or df.payment_status.isna().any():
        raise ValueError('PAYMENT STATUS FAIL — kategori di luar failed/successful/refunded')
    df['attempt_number']=pd.to_numeric(df.attempt_number,errors='coerce')
    df['payment_timestamp']=pd.to_datetime(df.payment_timestamp,errors='coerce')
    df['payment_amount']=pd.to_numeric(df.payment_amount,errors='coerce')
    if df[['attempt_number','payment_timestamp','payment_amount','payment_method']].isna().any().any():
        raise ValueError('PAYMENT INPUT FAIL — attempt/timestamp/amount/method kosong atau tidak valid')
    if (df.payment_amount<0).any(): raise ValueError('PAYMENT AMOUNT FAIL — nilai negatif perlu investigasi')
    if df.duplicated(['order_id','attempt_number']).any():
        raise ValueError('PAYMENT ATTEMPT FAIL — attempt_number duplikat dalam satu order')
    df=df.sort_values(['order_id','attempt_number','payment_timestamp','payment_id'],kind='stable')
    if (df.groupby('order_id')['payment_timestamp'].diff().dt.total_seconds().dropna()<0).any():
        raise ValueError('PAYMENT CHRONOLOGY FAIL — timestamp bertentangan dengan attempt_number')
    # Refund harus mempunyai successful attempt lebih dahulu pada order yang sama.
    # Hitung berdasarkan urutan attempt yang sudah divalidasi.
    success_seen=df.payment_status.eq('successful').groupby(df.order_id).cumsum()
    refund_without_success=df.payment_status.eq('refunded') & success_seen.eq(0)
    if refund_without_success.any():
        raise ValueError('PAYMENT REFUND FAIL — terdapat refund sebelum successful payment pada order yang sama')
    df['_success_amt']=np.where(df.payment_status.eq('successful'),df.payment_amount,0)
    df['_refund_amt']=np.where(df.payment_status.eq('refunded'),df.payment_amount,0)
    grp=df.groupby('order_id',as_index=False).agg(
        payment_attempt_count=('payment_id','size'),
        successful_payment_amount=('_success_amt','sum'),refunded_payment_amount=('_refund_amt','sum'))
    grp['net_payment_amount']=grp.successful_payment_amount-grp.refunded_payment_amount
    if (grp.net_payment_amount<0).any():
        raise ValueError('PAYMENT REFUND FAIL — refund melebihi successful amount')
    final=df.groupby('order_id',as_index=False).tail(1)[['order_id','payment_status']].rename(columns={'payment_status':'final_payment_status'})
    success=df.loc[df.payment_status.eq('successful')].groupby('order_id',as_index=False).tail(1)[['order_id','payment_method']]
    fallback=df.groupby('order_id',as_index=False).tail(1)[['order_id','payment_method']].rename(columns={'payment_method':'_last_method'})
    grp=grp.merge(final,on='order_id',how='left',validate='one_to_one')
    grp=grp.merge(success.rename(columns={'payment_method':'_success_method'}),on='order_id',how='left',validate='one_to_one')
    grp=grp.merge(fallback,on='order_id',how='left',validate='one_to_one')
    grp['payment_method_final']=grp['_success_method'].fillna(grp['_last_method'])
    return grp.drop(columns=['_success_method','_last_method'])

payment_agg=build_payment_aggregate(pm2_sources['Payments Clean']) if INPUT_FILES_READY else None
if payment_agg is not None:
    print('PAYMENT GRAIN:',aggregate_grain_check(payment_agg))
    print('Payments source rows:',len(pm2_sources['Payments Clean']), '| attempts preserved:',int(payment_agg.payment_attempt_count.sum()))
    display(payment_agg.head(5))
    display(payment_agg.final_payment_status.value_counts().rename('orders').to_frame())
else: print('PAYMENT AGG: PENDING')
```

Lihat `payment_attempt_count`, `successful_payment_amount`, `refunded_payment_amount`, `net_payment_amount`, `final_payment_status`, dan `payment_method_final`. Kriteria: total attempts aggregate = jumlah baris Payments source, key unik, net = successful − refunded. Jika fungsi menghentikan proses, jangan paksa PASS.

```python
# ★ LATIHAN OPSIONAL — boleh dilewati; bukan syarat penilaian.
# COBA — latihan perhitungan durasi dengan data delivered milik varianmu
# Tidak ada data asli yang diubah pada latihan ini.
if INPUT_FILES_READY:
    contoh_delivery = pm2_sources['Delivery'].loc[pm2_sources['Delivery'].delivery_status.eq('delivered')].head(1).copy()
    if not contoh_delivery.empty:
        awal = pd.to_datetime(contoh_delivery.iloc[0]['assigned_at'])
        akhir = pd.to_datetime(contoh_delivery.iloc[0]['delivered_at'])
        janji = float(contoh_delivery.iloc[0]['promised_delivery_minutes'])
        actual_minutes = (akhir-awal).total_seconds()/60
        print('Durasi aktual (menit):',actual_minutes,'| Durasi janji:',janji)
        print('Delay:',actual_minutes-janji,'| Terlambat?',actual_minutes>janji)
    else: print('Tidak ada record delivered; lanjutkan dan periksa source')
else: print('PENDING — Delivery belum terbaca')
```

```python
# ▶ WAJIB JALANKAN — kode utama siap pakai; jangan ubah fungsi.
def derive_delivery_features(delivery_df):
    out=delivery_df.copy(deep=True)
    required={'delivery_id','order_id','assigned_at','picked_up_at','delivered_at','delivery_status','promised_delivery_minutes'}
    if required-set(out): raise ValueError('DELIVERY COLUMNS MISSING: '+str(required-set(out)))
    for c in ['assigned_at','picked_up_at','delivered_at']:
        out[c]=pd.to_datetime(out[c],errors='coerce')
    out['promised_delivery_minutes']=pd.to_numeric(out['promised_delivery_minutes'],errors='coerce')
    delivered=out.delivery_status.eq('delivered')
    out['actual_delivery_minutes']=(out.delivered_at-out.assigned_at).dt.total_seconds().div(60).where(delivered)
    out['delay_minutes']=(out.actual_delivery_minutes-out.promised_delivery_minutes).where(delivered)
    out['is_late']=pd.Series(pd.NA,index=out.index,dtype='boolean')
    out.loc[delivered,'is_late']=(out.loc[delivered,'delay_minutes']>0).astype('boolean')
    return out

def delivery_lifecycle_check(df,orders_df=None):
    d=df.copy(deep=True)
    for c in ['assigned_at','picked_up_at','delivered_at']:
        d[c]=pd.to_datetime(d[c],errors='coerce')
    delivered=d.delivery_status.eq('delivered')
    invalid_status=~d.delivery_status.isin(['delivered','cancelled','incomplete'])
    missing_for_delivered=delivered & d[['assigned_at','picked_up_at','delivered_at']].isna().any(axis=1)
    order_violation=delivered & ((d.picked_up_at<d.assigned_at)|(d.delivered_at<d.picked_up_at))
    negative_promised=delivered & (pd.to_numeric(d.promised_delivery_minutes,errors='coerce').isna() | (pd.to_numeric(d.promised_delivery_minutes,errors='coerce')<0))
    cancelled_has_delivered=(~delivered)&d.delivered_at.notna()
    chronology_count=0
    if orders_df is not None:
        ref=orders_df[['order_id','order_timestamp']].copy()
        ref['order_timestamp']=pd.to_datetime(ref.order_timestamp,errors='coerce')
        tmp=d[['order_id','assigned_at']].merge(ref,on='order_id',how='left',validate='many_to_one')
        chronology_count=int((tmp.assigned_at<tmp.order_timestamp).sum())
    return {'invalid_status':int(invalid_status.sum()),'missing_delivered_timestamps':int(missing_for_delivered.sum()),
        'invalid_sequence':int(order_violation.sum()),'invalid_promised_minutes':int(negative_promised.sum()),
        'non_delivered_with_delivered_at':int(cancelled_has_delivered.sum()),'assigned_before_order':chronology_count}

delivery_features=derive_delivery_features(pm2_sources['Delivery']) if INPUT_FILES_READY else None
if delivery_features is not None:
    delivery_lifecycle_result=delivery_lifecycle_check(delivery_features,pm2_sources['Orders Clean'])
    print('DELIVERY LIFECYCLE:',delivery_lifecycle_result)
    display(delivery_features[['order_id','delivery_status','promised_delivery_minutes','actual_delivery_minutes','delay_minutes','is_late']].head(6))
else: print('DELIVERY FEATURES: PENDING')
```

## TAHAP 10 — Staged Joins ke Final Analytical Dataset

**CASE — Situasi SwiftBite:** Data yang sudah diringkas harus digabung tanpa menggandakan order.

**TUJUAN:** Menjalankan lima LEFT JOIN, memvalidasi key tabel kanan dan final grain.

**PAHAMI ISTILAH:** LEFT JOIN = pertahankan semua baris Orders; validate=one_to_one/many_to_one = pembatas hubungan saat join; fan-out = baris tergandakan.

**WORKING PAPER BARU:** Isi `02_Integrasi_dan_Rerun!C20:H24`: C join_key, D before_rows, E after_rows, F duplicate_order_rows_after, G status, H catatan unmatched. Ambil dari `integration_audit_df` / `wp_join_copy_df`. V-05/06, V-10–V-14 baru dicatat setelah Tahap 11.

> Ikuti cell dari atas ke bawah. Jalankan code cell dengan **Shift + Enter** atau tombol ▶ Run. *Markdown cell* adalah penjelasan yang tidak perlu dijalankan.

**LIHAT CONTOH:** Jika sebelum join Orders=100 dan sesudah join=110, ada potensi fan-out; jangan hapus 10 baris secara diam-diam. Cek uniqueness pada tabel kanan.

**COBA:** gunakan helper yang sudah siap pakai. Join bertahap sesuai blueprint. Jika fungsi menampilkan FAIL, periksa source yang disebutkan.

```python
# ▶ WAJIB JALANKAN — kode utama siap pakai; jangan ubah fungsi.
def join_with_audit(left,right,on,step_name,how='left'):
    if left is None or right is None: raise ValueError(step_name+': source belum tersedia')
    if on not in left or on not in right: raise ValueError(step_name+': join key '+on+' tidak ditemukan')
    if right[on].isna().any() or right[on].duplicated().any():
        raise ValueError(step_name+': right-side key tidak unik/null. Periksa grain sebelum join')
    before=len(left)
    unmatched=int((~left[on].isin(set(right[on]))).sum())
    out=left.merge(right,on=on,how=how,validate='many_to_one',suffixes=('','_right'))
    duplicate=int(out['order_id'].duplicated().sum())
    audit={'step':step_name,'join_key':on,'before_rows':before,'after_rows':len(out),
           'row_change':len(out)-before,'unmatched_left_rows':unmatched,
           'duplicate_order_rows_after':duplicate,'status':'PASS' if len(out)==before and duplicate==0 else 'FAIL'}
    if audit['status']!='PASS': raise ValueError('JOIN FAIL: '+str(audit))
    return out,audit

def build_integrated(orders_safe,customers_safe,restaurants_clean,item_agg,payment_agg,delivery_features):
    current=orders_safe.copy(deep=True)
    steps=[('Safe Customers',customers_safe,'customer_key'),
           ('Restaurants',restaurants_clean,'restaurant_id'),
           ('Item Aggregate',item_agg,'order_id'),
           ('Payment Aggregate',payment_agg,'order_id'),
           ('Delivery Features',delivery_features,'order_id')]
    audits=[]
    for label,right,key in steps:
        current,one_audit=join_with_audit(current,right,key,label)
        audits.append(one_audit)
    return current,pd.DataFrame(audits)

final_integrated,integration_audit_df=(build_integrated(orders_safe,customers_safe,pm2_sources['Restaurants Clean'],item_agg,payment_agg,delivery_features)
    if RUN_READY else (None,pd.DataFrame()))
if final_integrated is not None:
    display(integration_audit_df)
    print('FINAL GRAIN:',aggregate_grain_check(final_integrated))
    # Preview ONLY safe, non-PII sample; no raw customer_id
    display(final_integrated[['order_id','customer_key','item_count','payment_attempt_count','delivery_status']].head(5))
else: print('INTEGRATION: PENDING — input/salt/identitas belum lengkap')
```

**BANTUAN SALIN KE WORKING PAPER** — Jalankan code cell berikut agar nilai yang perlu dicatat dan nomor baris Excel tampil dalam satu tabel. Tabel ini tidak mengisi file Excel otomatis; peserta tetap membaca dan mencatat hasilnya.

```python
# ▶ WAJIB JALANKAN — kode utama siap pakai; jangan ubah fungsi.
REQUIRED_IDS=[f'V-{i:02d}' for i in range(1,20)]
VALIDATION_COLUMNS=['check_id','validation_level','dataset_relationship','validation_question','expected_criteria',
                    'actual_result_metric','evidence_output','status','related_stage','notes']
def validation_record(cid,level,source,question,expected,actual,evidence,ok,stage,notes=''):
    return dict(zip(VALIDATION_COLUMNS,[cid,level,source,question,expected,str(actual),str(evidence),
            'PASS' if bool(ok) else 'FAIL',stage,notes]))

def run_core_validations(src,cs,osafe,iagg,pagg,dfeat,final,audit,variant_id,salt):
    if any(x is None for x in [cs,osafe,iagg,pagg,dfeat,final,audit]):
        return pd.DataFrame(columns=VALIDATION_COLUMNS)
    customers=src['Customers Clean']; orders=src['Orders Clean'];items=src['Order Items Clean'];payments=src['Payments Clean'];rest=src['Restaurants Clean'];delivery=src['Delivery']
    rr=[]
    add=lambda cid,lev,s,q,exp,actual,evid,ok,stage:rr.append(validation_record(cid,lev,s,q,exp,actual,evid,ok,stage))
    forbidden=safe_schema_check(final)
    add('V-01','Safe Data','final_integrated','Direct PII absent?','No direct PII',forbidden,'safe_schema_check(final_integrated)',len(set(forbidden)&DIRECT_PII_FIELDS)==0,'PL-02')
    add('V-02','Safe Data','final_integrated','Raw customer_id absent?','customer_id absent','customer_id' in final.columns,'final.columns','customer_id' not in final.columns,'PL-02')
    rehash=customers.customer_id.map(lambda x:deterministic_hash(x,salt,variant_id))
    stable=rehash.equals(cs.customer_key.reset_index(drop=True))
    add('V-03','Safe Data','customer_key','Same input/salt same key?','identical on recompute',stable,'deterministic_hash + source IDs',stable,'PL-02')
    unmatched=fk_unmatched_count(osafe,cs,'customer_key','customer_key')
    add('V-04','Safe Data','Customers/Orders','Safe customer relation valid?','unmatched=0',unmatched,'fk_unmatched_count',unmatched==0 and cs.customer_key.is_unique,'PL-02')
    add('V-05','Grain','final_integrated','Final rows match Orders?','equal rows',f'{len(final)}/{len(orders)}','final_integrated + orders_pm1',len(final)==len(orders),'PL-06')
    grain=aggregate_grain_check(final)
    add('V-06','Grain','final_integrated','order_id unique and non-null?','duplicate=0 / missing=0',grain,'aggregate_grain_check',grain['valid'],'PL-06')
    ig=aggregate_grain_check(iagg)
    expected_item_ids=items.order_id.nunique()
    item_ok=ig['valid'] and len(iagg)==expected_item_ids and int(iagg.item_count.sum())==len(items) and np.isclose(iagg.items_subtotal.sum(),pd.to_numeric(items.line_amount).sum())
    add('V-07','Aggregation','Item Aggregate','1 row/order and conserved?','unique & item totals preserved',f'agg={len(iagg)}, unique source={expected_item_ids}, items={int(iagg.item_count.sum())}','item_agg + aggregate_grain_check',item_ok,'PL-03')
    pg=aggregate_grain_check(pagg)
    pay_ok=pg['valid'] and int(pagg.payment_attempt_count.sum())==len(payments) and len(pagg)==payments.order_id.nunique()
    add('V-08','Aggregation','Payment Aggregate','One row/order; attempts preserved?','unique & attempt count preserved',f'attempts={int(pagg.payment_attempt_count.sum())}/{len(payments)}','payment_agg + aggregate_grain_check',pay_ok,'PL-04')
    recomputed_success=payments.loc[payments.payment_status.eq('successful')].payment_amount.sum()
    recomputed_refund=payments.loc[payments.payment_status.eq('refunded')].payment_amount.sum()
    sums_ok=(np.isclose(pagg.successful_payment_amount.sum(),recomputed_success) and np.isclose(pagg.refunded_payment_amount.sum(),recomputed_refund)
             and np.allclose(pagg.net_payment_amount,pagg.successful_payment_amount-pagg.refunded_payment_amount)
             and (pagg.net_payment_amount>=0).all())
    # V-09 juga memeriksa setiap order. Total global yang sama tidak cukup:
    # kesalahan yang saling mengimbangi antar-order tetap harus terdeteksi.
    raw_money=payments.copy(deep=True)
    raw_money['_successful']=np.where(raw_money.payment_status.eq('successful'),raw_money.payment_amount,0)
    raw_money['_refunded']=np.where(raw_money.payment_status.eq('refunded'),raw_money.payment_amount,0)
    expected_money=raw_money.groupby('order_id',as_index=False).agg(
        expected_successful=('_successful','sum'), expected_refunded=('_refunded','sum'))
    money_check=expected_money.merge(
        pagg[['order_id','successful_payment_amount','refunded_payment_amount','net_payment_amount']],
        on='order_id', how='outer', indicator=True, validate='one_to_one')
    per_order_ok=bool(
        money_check['_merge'].eq('both').all() and
        np.isclose(money_check.expected_successful,money_check.successful_payment_amount).all() and
        np.isclose(money_check.expected_refunded,money_check.refunded_payment_amount).all() and
        np.isclose(money_check.net_payment_amount,
                   money_check.expected_successful-money_check.expected_refunded).all())
    sums_ok=bool(sums_ok and per_order_ok)
    add('V-09','Payment Logic','Payment Aggregate','Failed excluded/refund correct?','successful-refunded per order; failed excluded',
        f'success={recomputed_success}, refund={recomputed_refund}, per_order_match={per_order_ok}',
        'payment_agg vs payments_clean (per-order reconciliation)',sums_ok,'PL-04')
    add('V-10','Relationship','Safe Customers','Orders customer unmatched?','unmatched=0',unmatched,'safe relationship check',unmatched==0,'PL-06')
    unrest=fk_unmatched_count(orders,rest,'restaurant_id','restaurant_id')
    add('V-11','Relationship','Restaurants','Orders restaurant unmatched?','unmatched=0',unrest,'fk_unmatched_count',unrest==0,'PL-06')
    item_left=int((~orders.order_id.isin(set(iagg.order_id))).sum());item_audit=audit.loc[audit.step.eq('Item Aggregate')].iloc[0]
    add('V-12','Relationship','Item Aggregate','No fan-out and item coverage?','same rows; item coverage=100%',f'left_unmatched={item_left}, row_change={item_audit.row_change}','integration_audit_df',item_left==0 and item_audit.row_change==0,'PL-06')
    pay_left=int((~orders.order_id.isin(set(pagg.order_id))).sum());pay_audit=audit.loc[audit.step.eq('Payment Aggregate')].iloc[0]
    add('V-13','Relationship','Payment Aggregate','No fan-out and payment coverage?','same rows; payments covered',f'left_unmatched={pay_left}, row_change={pay_audit.row_change}','integration_audit_df',pay_left==0 and pay_audit.row_change==0,'PL-06')
    delivery_unmatched=int((~delivery.order_id.isin(set(orders.order_id))).sum());dg=aggregate_grain_check(dfeat)
    del_audit=audit.loc[audit.step.eq('Delivery Features')].iloc[0]
    add('V-14','Relationship','Delivery','Delivery valid 0..1/order?','key unique; no orphan; no fanout',f'delivery_orphan={delivery_unmatched}, delivery_rows={len(dfeat)}, row_change={del_audit.row_change}','delivery key + integration_audit_df',dg['valid'] and delivery_unmatched==0 and del_audit.row_change==0,'PL-06')
    lc=delivery_lifecycle_check(dfeat,orders)
    add('V-15','Delivery','Delivery lifecycle','Timestamp lifecycle valid?','all issue counts=0',lc,'delivery_lifecycle_check',all(int(v)==0 for v in lc.values()),'PL-05')
    delivered=dfeat.delivery_status.eq('delivered')
    expected_actual=(dfeat.delivered_at-dfeat.assigned_at).dt.total_seconds().div(60)
    actual_ok=np.allclose(dfeat.loc[delivered,'actual_delivery_minutes'].astype(float),expected_actual.loc[delivered].astype(float),equal_nan=False)
    delay_ok=np.allclose(dfeat.loc[delivered,'delay_minutes'].astype(float),
         (expected_actual-dfeat.promised_delivery_minutes).loc[delivered].astype(float),equal_nan=False)
    late_ok=bool((dfeat.loc[delivered,'is_late'].astype(bool).to_numpy()==(dfeat.loc[delivered,'delay_minutes']>0).to_numpy()).all())
    add('V-16','Delivery','Delivery features','Actual/delay/is_late correct?','formulas match',f'checked_delivered={int(delivered.sum())}, formula_matches={actual_ok and delay_ok and late_ok}','delivery_features + formula reconciliation',actual_ok and delay_ok and late_ok,'PL-05')
    non_delivered=~delivered
    null_ok=bool(dfeat.loc[non_delivered,['actual_delivery_minutes','delay_minutes','is_late']].isna().all().all())
    add('V-17','Delivery','Cancelled/incomplete','Three outcome fields null?','all null',f'non_delivered={int(non_delivered.sum())}, all_null={null_ok}','delivery_features null check',null_ok,'PL-05')
    return pd.DataFrame(rr,columns=VALIDATION_COLUMNS)

core_validation_df=(run_core_validations(pm2_sources,customers_safe,orders_safe,item_agg,payment_agg,delivery_features,final_integrated,integration_audit_df,VARIANT_ID,CONTROLLED_SALT)
    if RUN_READY else pd.DataFrame(columns=VALIDATION_COLUMNS))
if not core_validation_df.empty:
    display(core_validation_df[['check_id','validation_question','actual_result_metric','evidence_output','status']])
    print('CORE VALIDATION:',core_validation_df.status.value_counts().to_dict())
else: print('CORE VALIDATION: PENDING — tunggu input/salt lengkap')
```

## TAHAP 12 — Reproducibility — Tiga Controlled Runs

**CASE — Situasi SwiftBite:** Proses yang bergantung pada manual edit belum dapat dipakai ulang oleh tim Analytics.

**TUJUAN:** Menjalankan pipeline tiga kali dari input asal yang sama dan membandingkan signature.

**PAHAMI ISTILAH:** controlled run = input/parameter sama; signature = fingerprint hasil; reproducible = hasil konsisten tanpa edit di tengah.

**WORKING PAPER BARU:** Isi `02_Integrasi_dan_Rerun!B29:H33` (parameter, jangan tulis nilai salt), `B38:H40` (3 controlled runs), kesimpulan `A43:H45`. V-18 dan V-19 di `03_Validasi_dan_Keputusan!D29:G30`, berdasarkan `wp_rerun_validation_copy_df`.

> Ikuti cell dari atas ke bawah. Jalankan code cell dengan **Shift + Enter** atau tombol ▶ Run. *Markdown cell* adalah penjelasan yang tidak perlu dijalankan.

**LIHAT CONTOH:** RUN-01, RUN-02, RUN-03 masing-masing final_rows=N (N adalah Orders pada varianmu), 17/17 core PASS, signature sama.

**COBA:** jalankan fungsi pipeline di bawah. Fungsi harus memakai input asli (salinan data untuk tiap run) dan tidak menyunting DataFrame di tengah. Tiga run dijalankan dari file yang sama; jangan ubah salt atau identitas.

**Catatan:** signature dibandingkan sesudah mengurutkan baris dengan `order_id` dan kolom secara stabil. Ini menghindari perbedaan akibat urutan tampilan saja.

```python
# ▶ WAJIB JALANKAN — kode utama siap pakai; jangan ubah fungsi.
def dataframe_signature(df):
    if df is None: return None
    if 'order_id' not in df: raise ValueError('SIGNATURE: final data harus punya order_id')
    normalized=df.reindex(sorted(df.columns),axis=1).sort_values('order_id',kind='stable').reset_index(drop=True)
    for c in normalized.columns:
        if pd.api.types.is_datetime64_any_dtype(normalized[c]):
            normalized[c]=normalized[c].dt.strftime('%Y-%m-%dT%H:%M:%S.%f')
    payload=normalized.to_csv(index=False,na_rep='<NULL>',float_format='%.12g',lineterminator='\n').encode('utf-8')
    return hashlib.sha256(payload).hexdigest()

def source_input_signature(dataframes):
    # Fingerprint deterministik isi 6 input, diperiksa sebelum dan sesudah pipeline.
    h=hashlib.sha256()
    for name,df in sorted(dataframes.items()):
        h.update(name.encode('utf-8'))
        h.update(str(tuple(df.columns)).encode('utf-8'))
        h.update(str(tuple(str(d) for d in df.dtypes)).encode('utf-8'))
        h.update(df.to_csv(index=False,na_rep='<NULL>',float_format='%.12g',lineterminator='\n').encode('utf-8'))
    return h.hexdigest()

def run_pm2_pipeline(customers_input,orders_input,order_items_input,payments_input,restaurants_input,delivery_input,controlled_salt,variant_id):
    originals={'Customers Clean':customers_input,'Orders Clean':orders_input,
        'Order Items Clean':order_items_input,'Payments Clean':payments_input,
        'Restaurants Clean':restaurants_input,'Delivery':delivery_input}
    original_sig_before=source_input_signature(originals)
    src={'Customers Clean':customers_input.copy(deep=True),'Orders Clean':orders_input.copy(deep=True),
         'Order Items Clean':order_items_input.copy(deep=True),'Payments Clean':payments_input.copy(deep=True),
         'Restaurants Clean':restaurants_input.copy(deep=True),'Delivery':delivery_input.copy(deep=True)}
    cs,osafe=build_safe_customer_data(src['Customers Clean'],src['Orders Clean'],controlled_salt,variant_id)
    ia=build_item_aggregate(src['Order Items Clean'])
    pa=build_payment_aggregate(src['Payments Clean'])
    df=derive_delivery_features(src['Delivery'])
    final,audit=build_integrated(osafe,cs,src['Restaurants Clean'],ia,pa,df)
    val=run_core_validations(src,cs,osafe,ia,pa,df,final,audit,variant_id,controlled_salt)
    original_sig_after=source_input_signature(originals)
    input_unchanged=(original_sig_before==original_sig_after)
    if not input_unchanged:
        raise ValueError('RERUN INPUT FAIL — input sources berubah selama pipeline berjalan')
    return {'final_integrated':final,'integration_audit':audit,'core_validation':val,
        'run_summary':{'final_rows':len(final),'unique_order_id':int(final.order_id.nunique()),
           'core_pass_count':int(val.status.eq('PASS').sum()),'manual_intervention':not input_unchanged,'input_signature':original_sig_before,
           'output_signature':dataframe_signature(final)}}

pipeline_runs=[]; rerun_summary_df=pd.DataFrame(); rerun_validation_df=pd.DataFrame(columns=VALIDATION_COLUMNS)
if RUN_READY:
    args=[pm2_sources[k] for k in ['Customers Clean','Orders Clean','Order Items Clean','Payments Clean','Restaurants Clean','Delivery']]
    for j in range(1,4):
        result=run_pm2_pipeline(*args,CONTROLLED_SALT,VARIANT_ID)
        pipeline_runs.append(result)
    rerun_summary_df=pd.DataFrame([{'run_id':f'RUN-{i:02d}',**res['run_summary']} for i,res in enumerate(pipeline_runs,1)])
    signatures_same=rerun_summary_df.output_signature.nunique()==1
    inputs_same=rerun_summary_df.input_signature.nunique()==1
    rowcounts_same=rerun_summary_df.final_rows.nunique()==1 and rerun_summary_df.unique_order_id.nunique()==1
    core_all_pass=bool((rerun_summary_df.core_pass_count==17).all() and all(x['core_validation'].status.eq('PASS').all() for x in pipeline_runs))
    no_manual=bool((~rerun_summary_df.manual_intervention).all() and inputs_same)
    r18=validation_record('V-18','Reproducibility','Pipeline rerun','No manual intervention?','3 fresh reruns no manual intervention',
        f'runs={len(pipeline_runs)}, source_hashes_identical={inputs_same}, input_unchanged={no_manual}','rerun_summary_df',len(pipeline_runs)==3 and no_manual and core_all_pass,'PL-07')
    r19=validation_record('V-19','Reproducibility','Output signature','Signatures and counts identical?','same counts and SHA256 across runs',
        f'rows_consistent={rowcounts_same}, signature_consistent={signatures_same}','rerun_summary_df',signatures_same and rowcounts_same and core_all_pass,'PL-07')
    rerun_validation_df=pd.DataFrame([r18,r19],columns=VALIDATION_COLUMNS)
    display(rerun_summary_df.assign(output_signature=rerun_summary_df.output_signature.str.slice(0,16)+'…'))
    display(rerun_validation_df[['check_id','actual_result_metric','status']])
    # PS-03: rerun customer key check is also covered by same-salt signed output
else: print('RERUN: PENDING — lengkapi input/identitas/salt')
```

## TAHAP 13 — Final Pipeline Readiness Gate

**CASE — Situasi SwiftBite:** Tim Analytics hanya menerima final analytical dataset jika semua pemeriksaan wajib memiliki bukti dan input benar-benar dari varian yang terverifikasi.

**TUJUAN:** Memutuskan PASS/NOT READY/PENDING tanpa false PASS.

**PAHAMI ISTILAH:** gate = aturan keputusan; PENDING = belum lengkap; NOT READY = FAIL kritis; PASS = semua syarat terpenuhi.

**WORKING PAPER BARU:** Isi `03_Validasi_dan_Keputusan!D33` persis sama dengan `PM2_FINAL_GATE`. B33 mengambil Variant Check secara otomatis; F33:G33 memberi peringatan administrasi. Tulis alasan pada `A36:G39` dan masalah pada `A42:G44`.

> Ikuti cell dari atas ke bawah. Jalankan code cell dengan **Shift + Enter** atau tombol ▶ Run. *Markdown cell* adalah penjelasan yang tidak perlu dijalankan.

**PAHAMI:** Final Gate memeriksa 19 validation checks yang lengkap dan didukung evidence, serta input check teknis. Jika seluruhnya PASS, `PM2_FINAL_GATE=PASS`. Jika ada critical validation FAIL, status `NOT READY`. Jika evidence belum lengkap, status `PENDING`.

**PENTING:** Final Gate bersifat **kesiapan teknis pipeline**, bukan bukti independen bahwa peserta telah menerima paket varian yang benar. Kecocokan varian diverifikasi admin dari Handoff Tracker, tanpa file tambahan bagi peserta.


```python
# ▶ WAJIB JALANKAN — kode utama siap pakai; jangan ubah fungsi.
def derive_pm2_final_gate(df,source_status='PENDING',verified_validation=None):
    if source_status=='FAIL': return 'NOT READY'
    if df is None or len(df)==0: return 'PENDING'
    must={'check_id','status','actual_result_metric','evidence_output'}
    if must-set(df.columns): return 'PENDING'
    # Gate menerima bukti yang sama dengan pemeriksaan ulang dari data aktual.
    if verified_validation is None or not df.equals(verified_validation): return 'PENDING'
    ids=df.check_id.astype(str)
    if len(df)!=19 or ids.nunique()!=19 or set(ids)!=set(REQUIRED_IDS): return 'PENDING'
    status=df.status.astype(str).str.upper()
    if not status.isin(['PASS','FAIL','PENDING']).all(): return 'PENDING'
    if status.eq('FAIL').any(): return 'NOT READY'
    evidence_complete=bool(df[['actual_result_metric','evidence_output']].apply(
        lambda col: col.notna() & col.astype(str).str.strip().ne('')).all().all())
    if not evidence_complete or not status.eq('PASS').all() or source_status!='PASS': return 'PENDING'
    return 'PASS'

validation_df=pd.concat([core_validation_df,rerun_validation_df],ignore_index=True)
verified_core_validation=(run_core_validations(pm2_sources,customers_safe,orders_safe,item_agg,
    payment_agg,delivery_features,final_integrated,integration_audit_df,VARIANT_ID,CONTROLLED_SALT)
    if RUN_READY else pd.DataFrame(columns=VALIDATION_COLUMNS))
verified_validation_df=pd.concat([verified_core_validation,rerun_validation_df],ignore_index=True)
PM2_FINAL_GATE=derive_pm2_final_gate(validation_df,input_check_result.get('status'),
    verified_validation=verified_validation_df)
print('VALIDATION STATUS:',validation_df.status.value_counts().to_dict() if not validation_df.empty else {})
print('SOURCE CHECK:',input_check_result.get('status'),'|',input_check_result.get('reason'))
print('PM2 PIPELINE READINESS GATE:',PM2_FINAL_GATE)
print('CATATAN: kecocokan paket dengan Variant ID diverifikasi admin melalui Handoff Tracker.')
if PM2_FINAL_GATE=='PENDING': print('ACTION: lengkapi seluruh input/evidence serta check IDs yang masih PENDING.')
elif PM2_FINAL_GATE=='NOT READY': print('ACTION: kembali ke tahap yang FAIL, perbaiki dan rerun.')
else: print('ACTION: siap final export, lalu verifikasi hasil.')
FINAL_JUDGMENT_RATIONALE=(f'{len(validation_df)} checks recorded, '+
    f"{int(validation_df.status.eq('PASS').sum()) if not validation_df.empty else 0} PASS; "+
    f'Input technical check={input_check_result.get("status")}; final gate={PM2_FINAL_GATE}. '+
    ('Semua syarat teknis final tervalidasi; identitas varian diperiksa admin.' if PM2_FINAL_GATE=='PASS' else 'Catat issue dan follow-up sesuai output.'))
print('Rationale draft:',FINAL_JUDGMENT_RATIONALE)
```

**ISI WORKING PAPER:** Isi `03_Validasi_dan_Keputusan!D33` persis seperti `PM2_FINAL_GATE`. B33 otomatis mengambil Source Check dari sheet 01. Tulis alasan di `A36:G39` dan masalah di `A42:G44`.

## TAHAP 14 — Export Evidence & LMS Submission

**CASE — Situasi SwiftBite:** Mentor perlu memeriksa hasil yang dapat ditelusuri. File evidence boleh tersedia ketika pipeline PENDING, tetapi tidak boleh diklaim sebagai final approved dataset.

**TUJUAN:** Mengekspor dokumen teknis ke folder yang jelas serta memeriksa kelengkapan sebelum upload.

**PAHAMI ISTILAH:** export = menyimpan hasil; traceability = jejak bukti; approved output = data final yang memenuhi seluruh gate.

**WORKING PAPER BARU:** Cek seluruh sheet 01–03 lalu ikuti ketentuan LMS. Bukti hanya dikirim sesuai Participant Brief; jangan mengunggah raw source, PII, atau controlled salt.

> Ikuti cell dari atas ke bawah. Jalankan code cell dengan **Shift + Enter** atau tombol ▶ Run. *Markdown cell* adalah penjelasan yang tidak perlu dijalankan.

**PERHATIAN:** Output yang dibuat otomatis berbeda dari Working Paper yang harus kamu isi sendiri. Jangan upload raw dataset PM1, Delivery raw, atau salt kecuali admin LMS secara eksplisit meminta.

**COBA:** jalankan export. File evidence dapat diekspor walau gate belum PASS. `swiftbite_pm2_integrated_order_level.csv` **hanya** dibuat sebagai approved final output jika gate PASS. Jika run terbaru bukan PASS, Notebook akan **memindahkan CSV final dari run sebelumnya** ke folder arsip lokal (`outputs/pm2_swiftbite/_arsip_bukan_hasil_final/`), bukan menghapusnya. **Jangan masukkan file arsip ke LMS.**

**ISI WORKING PAPER BARU:** Setelah export, periksa `03_Validasi_dan_Keputusan!D33`, `A36:G39`, `A42:G44` dan kesimpulan rerun di `02_Integrasi_dan_Rerun!A43:H45`. Tidak ada lagi baris PL-08 khusus pada format Working Paper ringkas.

**LANGKAH SIMPAN TANPA:** setelah cell export selesai, buka Explorer/File Manager pada folder utama → `outputs/pm2_swiftbite/` untuk tiga evidence. Jika `PM2_FINAL_GATE=PASS`, cek juga `data/processed/pm2_swiftbite/swiftbite_pm2_integrated_order_level.csv`. Simpan Notebook dengan **Ctrl+S**, kemudian simpan Working Paper Excel. Ikuti petunjuk ZIP/upload pada LMS. **Jika gate belum PASS, jangan masukkan CSV final lama** meskipun pernah ada di folder dari percobaan sebelumnya.


```python
# ▶ WAJIB JALANKAN — kode utama siap pakai; jangan ubah fungsi.
def export_pm2_evidence(audit,validation,summary,output_dir):
    output_dir.mkdir(parents=True,exist_ok=True)
    audit.to_csv(output_dir/'pm2_integration_audit.csv',index=False)
    validation.to_csv(output_dir/'pm2_validation_results.csv',index=False)
    (output_dir/'pm2_pipeline_readiness_summary.json').write_text(json.dumps(summary,ensure_ascii=False,indent=2,default=str),encoding='utf-8')
    return [output_dir/'pm2_integration_audit.csv',output_dir/'pm2_validation_results.csv',output_dir/'pm2_pipeline_readiness_summary.json']

def archive_previous_unapproved_final(processed_dir, archive_root):
    """Amankan file final lama saat run TERBARU tidak PASS (tanpa menghapus evidence)."""
    target = processed_dir / 'swiftbite_pm2_integrated_order_level.csv'
    if not target.is_file():
        return None
    from datetime import datetime, timezone
    archive_root.mkdir(parents=True, exist_ok=True)
    stamp = datetime.now(timezone.utc).strftime('%Y%m%dT%H%M%S%fZ')
    archived = archive_root / f'swiftbite_pm2_integrated_order_level_ARSIP_{stamp}.csv'
    target.replace(archived)
    print('FILE FINAL RUN LAMA DIPINDAHKAN KE ARSIP (BUKAN UNTUK LMS).')
    print('Lokasi arsip:', archived)
    return archived

def export_final_dataset(final_df,final_gate,processed_dir,archive_root):
    if final_gate!='PASS':
        archive_previous_unapproved_final(processed_dir, archive_root)
        print('FINAL CSV NOT EXPORTED: gate bukan PASS. Tidak ada CSV approved di folder final.')
        return None
    if final_df is None or final_df.empty:
        raise ValueError('FINAL EXPORT: final dataset kosong')
    if safe_schema_check(final_df):
        raise ValueError('FINAL EXPORT: ada PII atau raw customer ID')
    processed_dir.mkdir(parents=True,exist_ok=True)
    target=processed_dir/'swiftbite_pm2_integrated_order_level.csv'
    # Simpan ke temp file dulu agar export yang gagal tidak merusak CSV sebelumnya.
    import tempfile
    tmp_path = None
    try:
        with tempfile.NamedTemporaryFile(mode='w', encoding='utf-8', newline='',
                                         suffix='.csv', prefix='.pm2_export_',
                                         dir=processed_dir, delete=False) as handle:
            tmp_path = Path(handle.name)
            final_df.to_csv(handle, index=False)
        os.replace(tmp_path, target)
    finally:
        if tmp_path is not None and tmp_path.exists():
            tmp_path.unlink()
    return target

if not validation_df.empty:
    summary={'participant_name':PARTICIPANT_NAME,'variant_id':VARIANT_ID,'input_check_status':input_check_result.get('status'),
        'variant_handoff_control':'ADMIN_HANDOFF_TRACKER','pm2_final_gate':PM2_FINAL_GATE,
        'validation_pass_count':int(validation_df.status.eq('PASS').sum()),
        'validation_fail_count':int(validation_df.status.eq('FAIL').sum()),
        'rationale':FINAL_JUDGMENT_RATIONALE,
        'output_signature':dataframe_signature(final_integrated) if final_integrated is not None else None}
    final_path=export_final_dataset(final_integrated,PM2_FINAL_GATE,PM2_PROCESSED_DIR,PM2_OUTPUT_DIR/'_arsip_bukan_hasil_final')
    summary['final_dataset_export_status'] = 'APPROVED' if final_path else 'NOT_EXPORTED'
    evidence_paths=export_pm2_evidence(integration_audit_df,validation_df,summary,PM2_OUTPUT_DIR)
    print('EVIDENCE EXPORTED:')
    for path in evidence_paths: print(' -',path)
    print('FINAL DATASET:', final_path or 'Not exported (PENDING/NOT READY)')
else:
    archive_previous_unapproved_final(PM2_PROCESSED_DIR,PM2_OUTPUT_DIR/'_arsip_bukan_hasil_final')
    print('EXPORT: PENDING — validation belum tersedia. CSV final lama (jika ada) dipindahkan ke arsip.')
```
